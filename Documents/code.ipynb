```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
"""
================================================================================
Neural Differential Equations for Wear Failure Dynamics:
A Physics-Informed Framework for Discovering and Solving Degradation Laws
in Machining Systems
================================================================================
Authors: Muhammad Umar, Faisal Saleem, Jong-Myon Kim
Affiliation: UIAI Lab, University of Ulsan, South Korea

Target: Mechanical Systems and Signal Processing (MSSP) /
        Reliability Engineering & System Safety (RESS)

Hardware: NVIDIA GeForce GTX 1080 (8GB VRAM)
================================================================================
"""

# %% [markdown]
# # Cell 1: Environment Setup & Imports

# %%
import os
import sys
import glob
import warnings
import random
import time
import math
from pathlib import Path
from collections import OrderedDict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, TensorDataset
from torch.optim.lr_scheduler import CosineAnnealingWarmRestarts, ReduceLROnPlateau

from scipy import signal as scipy_signal
from scipy.stats import pearsonr, spearmanr, kendalltau
from scipy.interpolate import interp1d
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

import pywt

warnings.filterwarnings("ignore")
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 14,
    'axes.titlesize': 14,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11,
    'figure.dpi': 150,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'font.family': 'serif',
    'mathtext.fontset': 'cm',
})

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"Device: {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")

# %% [markdown]
# # Cell 2: Dataset Configuration & Loading

# %%
# ========= DATASET CONFIG =========
BASE = r"E:\Collaboration Work\With Farooq\phm dataset\PHM Challange 2010 Milling"
TRAIN_CUTTERS = ["c1", "c4", "c6"]
TEST_CUTTERS  = ["c2", "c3", "c5"]

SENSOR_COLS = ["Force_X", "Force_Y", "Force_Z",
               "Vibration_X", "Vibration_Y", "Vibration_Z",
               "AE_RMS"]
NUM_SENSORS = len(SENSOR_COLS)
SAMPLING_RATE = 50000  # 50 kHz
NUM_CUTS = 315
NUM_FLUTES = 3

# Hyperparameters
FEATURE_DIM = 84          # Features per cut after extraction
HIDDEN_DIM = 128          # Neural ODE hidden dimension
LATENT_DIM = 64           # Latent state dimension
NUM_EPOCHS = 300
BATCH_SIZE = 16
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-5
SEQ_LEN = 20             # Sliding window length for sequential modeling
PRED_HORIZON = 1          # Predict next step


def load_cut_data(cutter_id, cut_num, base_path=BASE):
    """Load raw sensor data for a single cut."""
    filepath = os.path.join(base_path, cutter_id, cutter_id,
                            f"c_{cutter_id[1]}_{cut_num:03d}.csv")
    df = pd.read_csv(filepath, header=None)
    df.columns = SENSOR_COLS
    return df


def load_wear_data(cutter_id, base_path=BASE):
    """Load wear measurements for a cutter."""
    filepath = os.path.join(base_path, cutter_id, f"{cutter_id}_wear.csv")
    df = pd.read_csv(filepath)
    return df


# Verify data loading
print("=" * 60)
print("DATA LOADING VERIFICATION")
print("=" * 60)
for cid in TRAIN_CUTTERS:
    wear = load_wear_data(cid)
    sample_cut = load_cut_data(cid, 1)
    print(f"\n{cid.upper()}: {wear.shape[0]} cuts, "
          f"Wear range: [{wear['flute_1'].min():.1f}, {wear['flute_1'].max():.1f}] μm")
    print(f"  Sample cut shape: {sample_cut.shape}, "
          f"Duration: {sample_cut.shape[0]/SAMPLING_RATE*1000:.1f} ms")

# %% [markdown]
# # Cell 3: Multi-Domain Feature Extraction Engine
# Extracts time-domain, frequency-domain, time-frequency, and physics-informed features.

# %%
class MultiDomainFeatureExtractor:
    """
    Comprehensive multi-domain feature extraction for machining signals.
    Extracts 12 features per sensor channel = 84 total features per cut.

    Domains:
        - Time domain: statistical + nonlinear features
        - Frequency domain: spectral features
        - Time-frequency domain: wavelet energy features
        - Physics-informed: cutting mechanics features
    """

    def __init__(self, fs=SAMPLING_RATE):
        self.fs = fs

    def time_domain_features(self, x):
        """Extract time-domain statistical features."""
        rms = np.sqrt(np.mean(x**2))
        peak = np.max(np.abs(x))
        crest = peak / (rms + 1e-10)
        kurtosis = float(pd.Series(x).kurtosis())
        skewness = float(pd.Series(x).skew())
        std = np.std(x)
        mean_abs = np.mean(np.abs(x))
        shape_factor = rms / (mean_abs + 1e-10)
        return [rms, peak, crest, kurtosis, skewness, std, mean_abs, shape_factor]

    def frequency_domain_features(self, x):
        """Extract frequency-domain spectral features."""
        n = len(x)
        fft_vals = np.abs(np.fft.rfft(x))
        freqs = np.fft.rfftfreq(n, d=1.0/self.fs)

        # Spectral centroid
        total_power = np.sum(fft_vals**2) + 1e-10
        spectral_centroid = np.sum(freqs * fft_vals**2) / total_power

        # Spectral bandwidth
        spectral_bw = np.sqrt(np.sum((freqs - spectral_centroid)**2 * fft_vals**2) / total_power)

        return [spectral_centroid, spectral_bw]

    def wavelet_energy_features(self, x, wavelet='db4', level=4):
        """Extract wavelet packet energy distribution."""
        coeffs = pywt.wavedec(x, wavelet, level=level)
        energies = [np.sum(c**2) for c in coeffs]
        total = sum(energies) + 1e-10
        # Return normalized energy of detail coefficients at 2 highest levels
        return [energies[1]/total, energies[2]/total]

    def extract_single_channel(self, x):
        """Extract all features for a single sensor channel."""
        tf = self.time_domain_features(x)        # 8 features
        ff = self.frequency_domain_features(x)    # 2 features
        wf = self.wavelet_energy_features(x)      # 2 features
        return tf + ff + wf                        # 12 features total

    def extract_cut_features(self, cut_df):
        """Extract features for all channels of a single cut."""
        features = []
        for col in SENSOR_COLS:
            x = cut_df[col].values
            features.extend(self.extract_single_channel(x))
        return np.array(features, dtype=np.float32)

    def extract_cutter_features(self, cutter_id, n_cuts=NUM_CUTS):
        """Extract features for all cuts of a cutter."""
        print(f"  Extracting features for {cutter_id}...", end=" ")
        all_features = []
        for i in range(1, n_cuts + 1):
            cut_df = load_cut_data(cutter_id, i)
            feat = self.extract_cut_features(cut_df)
            all_features.append(feat)
            if i % 50 == 0:
                print(f"{i}", end=" ")
        print("Done!")
        return np.array(all_features)  # (N_cuts, FEATURE_DIM)


# Extract features for all cutters
extractor = MultiDomainFeatureExtractor()

print("\n" + "=" * 60)
print("FEATURE EXTRACTION")
print("=" * 60)

train_features = {}
train_wear = {}
for cid in TRAIN_CUTTERS:
    train_features[cid] = extractor.extract_cutter_features(cid)
    train_wear[cid] = load_wear_data(cid)
    print(f"  {cid}: Features shape = {train_features[cid].shape}")

test_features = {}
for cid in TEST_CUTTERS:
    test_features[cid] = extractor.extract_cutter_features(cid)
    print(f"  {cid}: Features shape = {test_features[cid].shape}")

print(f"\nTotal features per cut: {train_features['c1'].shape[1]}")

# %% [markdown]
# # Cell 4: Feature Normalization & Wear Target Preparation

# %%
# ========= NORMALIZATION =========
# Fit scaler on training data only
all_train_feats = np.vstack([train_features[cid] for cid in TRAIN_CUTTERS])
feature_scaler = StandardScaler()
feature_scaler.fit(all_train_feats)

# Normalize features
for cid in TRAIN_CUTTERS:
    train_features[cid] = feature_scaler.transform(train_features[cid])
for cid in TEST_CUTTERS:
    test_features[cid] = feature_scaler.transform(test_features[cid])

# ========= WEAR TARGETS =========
# Compute total wear (average of 3 flutes) - primary target
# Also keep individual flute wear for multi-task learning
wear_scaler = MinMaxScaler(feature_range=(0, 1))

all_train_total_wear = []
for cid in TRAIN_CUTTERS:
    w = train_wear[cid]
    total_w = w[['flute_1', 'flute_2', 'flute_3']].mean(axis=1).values
    all_train_total_wear.append(total_w)

all_train_total_wear_arr = np.concatenate(all_train_total_wear).reshape(-1, 1)
wear_scaler.fit(all_train_total_wear_arr)

train_wear_normalized = {}
train_flute_wear_normalized = {}
for cid in TRAIN_CUTTERS:
    w = train_wear[cid]
    total_w = w[['flute_1', 'flute_2', 'flute_3']].mean(axis=1).values.reshape(-1, 1)
    train_wear_normalized[cid] = wear_scaler.transform(total_w).flatten()

    # Individual flute wear (normalized by same scaler for consistency)
    flute_w = w[['flute_1', 'flute_2', 'flute_3']].values
    train_flute_wear_normalized[cid] = wear_scaler.transform(
        flute_w.reshape(-1, 1)).reshape(-1, 3)

# Compute RUL targets (normalized 1->0)
train_rul = {}
for cid in TRAIN_CUTTERS:
    n = len(train_wear_normalized[cid])
    train_rul[cid] = np.linspace(1.0, 0.0, n).astype(np.float32)

print("=" * 60)
print("DATA PREPARATION SUMMARY")
print("=" * 60)
for cid in TRAIN_CUTTERS:
    print(f"{cid}: Features {train_features[cid].shape}, "
          f"Wear range [{train_wear_normalized[cid].min():.3f}, "
          f"{train_wear_normalized[cid].max():.3f}], "
          f"RUL range [{train_rul[cid].min():.3f}, {train_rul[cid].max():.3f}]")

# %% [markdown]
# # Cell 5: Sequential Dataset Construction (Sliding Window)

# %%
class ToolWearSequenceDataset(Dataset):
    """
    Creates overlapping sequences for temporal modeling.
    Each sample: (feature_seq, wear_seq, rul_seq, flute_wear_seq, time_indices)
    """

    def __init__(self, features, wear, rul, flute_wear, seq_len=SEQ_LEN):
        self.seq_len = seq_len
        self.samples = []

        n = len(features)
        for i in range(n - seq_len):
            feat_seq = features[i:i+seq_len]
            wear_seq = wear[i:i+seq_len]
            rul_seq = rul[i:i+seq_len]
            flute_seq = flute_wear[i:i+seq_len]
            time_idx = np.arange(i, i+seq_len, dtype=np.float32) / n  # Normalized time

            self.samples.append((
                torch.FloatTensor(feat_seq),
                torch.FloatTensor(wear_seq),
                torch.FloatTensor(rul_seq),
                torch.FloatTensor(flute_seq),
                torch.FloatTensor(time_idx),
            ))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return self.samples[idx]


# Build datasets
train_datasets = []
for cid in TRAIN_CUTTERS:
    ds = ToolWearSequenceDataset(
        train_features[cid],
        train_wear_normalized[cid],
        train_rul[cid],
        train_flute_wear_normalized[cid],
        seq_len=SEQ_LEN
    )
    train_datasets.append(ds)
    print(f"{cid}: {len(ds)} sequences of length {SEQ_LEN}")

combined_train_ds = torch.utils.data.ConcatDataset(train_datasets)
train_loader = DataLoader(combined_train_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=0, pin_memory=True, drop_last=True)

print(f"\nTotal training sequences: {len(combined_train_ds)}")
print(f"Batches per epoch: {len(train_loader)}")

# %% [markdown]
# # Cell 6: Physics-Informed Neural ODE Components
# Core architecture: Neural ODE with physics-constrained wear dynamics

# %%
# ========================
# 6A: ODE Function Network
# ========================
class WearRateODE(nn.Module):
    """
    Neural ODE function: dW/dt = f_physics(W, t) + f_neural(z, W, t)

    The wear rate is decomposed into:
    - Physics component: Monotonic increasing rate based on Archard/Usui principles
    - Neural residual: Learned correction capturing unknown tribological effects

    Key constraint: dW/dt >= 0 (wear is irreversible)
    """

    def __init__(self, latent_dim, hidden_dim=128):
        super().__init__()
        self.latent_dim = latent_dim

        # Physics-inspired wear rate (parameterized Archard-like model)
        # dW/dt_physics = a * exp(b * W) * (1 + c * t)
        self.physics_a = nn.Parameter(torch.tensor(0.1))
        self.physics_b = nn.Parameter(torch.tensor(0.01))
        self.physics_c = nn.Parameter(torch.tensor(0.1))

        # Neural residual network
        self.neural_net = nn.Sequential(
            nn.Linear(latent_dim + 2, hidden_dim),  # +2 for W and t
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, 1),
        )

        # Softplus to ensure non-negative wear rate
        self.softplus = nn.Softplus(beta=5)

    def physics_rate(self, W, t):
        """Archard-Usui inspired physics wear rate."""
        a = self.softplus(self.physics_a)
        b = self.softplus(self.physics_b)
        c = self.softplus(self.physics_c)
        rate = a * torch.exp(b * W) * (1.0 + c * t)
        return rate

    def forward(self, t_val, state):
        """
        Args:
            t_val: scalar time (normalized)
            state: (batch, latent_dim + 1) where last dim is W
        Returns:
            d_state/dt: (batch, latent_dim + 1)
        """
        z = state[:, :-1]        # Latent state: (batch, latent_dim)
        W = state[:, -1:]        # Wear state: (batch, 1)

        t_input = t_val * torch.ones_like(W)

        # Physics rate
        phys_rate = self.physics_rate(W, t_input)

        # Neural residual
        neural_input = torch.cat([z, W, t_input], dim=-1)
        neural_rate = self.neural_net(neural_input)

        # Total wear rate (guaranteed non-negative)
        total_rate = self.softplus(phys_rate + neural_rate)

        # Latent dynamics (unconstrained)
        dz = torch.zeros_like(z)  # Latent state evolves through encoder, not ODE

        dW = total_rate
        d_state = torch.cat([dz, dW], dim=-1)
        return d_state


# ========================
# 6B: Simple Euler ODE Solver (Memory efficient for GTX 1080)
# ========================
class EulerODESolver(nn.Module):
    """
    Fixed-step Euler solver for Neural ODE.
    Memory efficient alternative to adaptive solvers.
    """

    def __init__(self, ode_func, n_steps=10):
        super().__init__()
        self.ode_func = ode_func
        self.n_steps = n_steps

    def forward(self, initial_state, t_span):
        """
        Solve ODE from t_span[0] to t_span[1].

        Args:
            initial_state: (batch, state_dim)
            t_span: (2,) tensor [t0, t1]
        Returns:
            final_state: (batch, state_dim)
        """
        t0, t1 = t_span[0], t_span[1]
        dt = (t1 - t0) / self.n_steps
        state = initial_state

        for i in range(self.n_steps):
            t_val = t0 + i * dt
            d_state = self.ode_func(t_val, state)
            state = state + dt * d_state

        return state


print("ODE components defined successfully.")
print(f"  WearRateODE: Physics + Neural hybrid wear dynamics")
print(f"  EulerODESolver: {10}-step fixed solver (GTX 1080 optimized)")

# %% [markdown]
# # Cell 7: Complete PINN Framework Architecture

# %%
class PhysicsInformedNeuralWearModel(nn.Module):
    """
    ===================================================================
    PINN-NDE: Physics-Informed Neural Differential Equation Framework
    for Tool Wear Degradation Modeling
    ===================================================================

    Architecture:
    1. Feature Encoder: Maps raw features to latent representation
    2. Temporal Encoder: GRU-based sequence modeling
    3. Neural ODE: Physics-constrained wear evolution
    4. Multi-Head Decoder: Wear, RUL, and flute-level predictions
    5. Uncertainty Head: Aleatoric + epistemic uncertainty

    Physics Constraints (enforced in loss):
    - Monotonicity: dW/dt >= 0 (irreversible degradation)
    - Energy consistency: Work ∝ material removal + friction
    - Archard proportionality: W ∝ F·L/H
    - Phase-aware dynamics: Different ODE branches for wear phases
    """

    def __init__(self, input_dim=FEATURE_DIM, hidden_dim=HIDDEN_DIM,
                 latent_dim=LATENT_DIM, n_flutes=NUM_FLUTES, ode_steps=10):
        super().__init__()

        self.latent_dim = latent_dim
        self.hidden_dim = hidden_dim

        # ===== 1. Feature Encoder =====
        self.feature_encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.SiLU(),
            nn.Dropout(0.1),
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, latent_dim),
        )

        # ===== 2. Temporal Encoder (Bidirectional GRU) =====
        self.temporal_encoder = nn.GRU(
            input_size=latent_dim,
            hidden_size=hidden_dim // 2,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=0.1,
        )
        self.temporal_proj = nn.Linear(hidden_dim, latent_dim)

        # ===== 3. Neural ODE =====
        self.ode_func = WearRateODE(latent_dim, hidden_dim)
        self.ode_solver = EulerODESolver(self.ode_func, n_steps=ode_steps)

        # Initial wear state predictor
        self.initial_wear_net = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, 1),
            nn.Sigmoid(),  # Normalized wear [0, 1]
        )

        # ===== 4. Multi-Head Decoder =====
        # Total wear decoder
        self.wear_decoder = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim // 2),  # +1 for ODE wear state
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, 1),
            nn.Sigmoid(),
        )

        # Flute-level wear decoder
        self.flute_decoder = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, n_flutes),
            nn.Sigmoid(),
        )

        # RUL decoder
        self.rul_decoder = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, 1),
            nn.Sigmoid(),
        )

        # ===== 5. Uncertainty Head =====
        self.uncertainty_head = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim // 4),
            nn.SiLU(),
            nn.Linear(hidden_dim // 4, 2),  # log_var for wear and RUL
        )

        # ===== 6. Phase Classifier =====
        # Classifies wear phase: 0=initial, 1=steady, 2=accelerated
        self.phase_classifier = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim // 4),
            nn.SiLU(),
            nn.Linear(hidden_dim // 4, 3),
        )

        # ===== 7. Latent Physics Inference Head =====
        # Infers latent temperature and stress from sensor features
        self.latent_physics = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 4),
            nn.SiLU(),
            nn.Linear(hidden_dim // 4, 2),  # [temperature_proxy, stress_proxy]
            nn.Softplus(),  # Non-negative physical quantities
        )

    def forward(self, feat_seq, time_seq):
        """
        Forward pass through the complete framework.

        Args:
            feat_seq: (batch, seq_len, input_dim) - Feature sequences
            time_seq: (batch, seq_len) - Normalized time indices

        Returns:
            dict with all predictions and intermediate states
        """
        batch_size, seq_len, _ = feat_seq.shape

        # 1. Encode features
        encoded = self.feature_encoder(feat_seq)  # (B, T, latent_dim)

        # 2. Temporal encoding
        temporal_out, _ = self.temporal_encoder(encoded)  # (B, T, hidden_dim)
        z = self.temporal_proj(temporal_out)  # (B, T, latent_dim)

        # 3. Neural ODE for wear evolution
        # Get initial wear state from first timestep
        W0 = self.initial_wear_net(z[:, 0, :])  # (B, 1)

        # Solve ODE step by step through the sequence
        wear_states = [W0]
        physics_rates = []
        current_W = W0

        for t in range(1, seq_len):
            # Build state: [latent, wear]
            state = torch.cat([z[:, t-1, :], current_W], dim=-1)

            t_start = time_seq[:, t-1].mean()
            t_end = time_seq[:, t].mean()
            t_span = torch.tensor([t_start, t_end], device=feat_seq.device)

            # Solve ODE
            next_state = self.ode_solver(state, t_span)
            current_W = next_state[:, -1:]

            wear_states.append(current_W)

            # Track physics rate for analysis
            with torch.no_grad():
                pr = self.ode_func.physics_rate(current_W, t_end * torch.ones_like(current_W))
                physics_rates.append(pr)

        # Stack wear states: (B, T, 1)
        wear_ode = torch.stack(wear_states, dim=1)

        # 4. Decode predictions at each timestep
        decoder_input = torch.cat([z, wear_ode], dim=-1)  # (B, T, latent+1)

        wear_pred = self.wear_decoder(decoder_input).squeeze(-1)     # (B, T)
        flute_pred = self.flute_decoder(decoder_input)               # (B, T, 3)
        rul_pred = self.rul_decoder(decoder_input).squeeze(-1)       # (B, T)

        # 5. Uncertainty estimation
        uncertainty = self.uncertainty_head(decoder_input)  # (B, T, 2)
        log_var_wear = uncertainty[:, :, 0]
        log_var_rul = uncertainty[:, :, 1]

        # 6. Phase classification
        phase_logits = self.phase_classifier(decoder_input)  # (B, T, 3)

        # 7. Latent physics inference
        latent_phys = self.latent_physics(z)  # (B, T, 2)

        return {
            'wear_pred': wear_pred,
            'flute_pred': flute_pred,
            'rul_pred': rul_pred,
            'wear_ode': wear_ode.squeeze(-1),
            'log_var_wear': log_var_wear,
            'log_var_rul': log_var_rul,
            'phase_logits': phase_logits,
            'latent_temperature': latent_phys[:, :, 0],
            'latent_stress': latent_phys[:, :, 1],
            'latent_z': z,
            'physics_rates': physics_rates,
        }


# Instantiate model
model = PhysicsInformedNeuralWearModel().to(DEVICE)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\n{'='*60}")
print(f"MODEL ARCHITECTURE SUMMARY")
print(f"{'='*60}")
print(f"Total parameters:     {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Model size:           {total_params * 4 / 1e6:.2f} MB (float32)")
print(f"{'='*60}")

# %% [markdown]
# # Cell 8: Physics-Informed Loss Function

# %%
class PhysicsInformedLoss(nn.Module):
    """
    Multi-objective physics-informed loss function.

    Components:
    1. Data fidelity: MSE for wear, RUL, and flute-level predictions
    2. Monotonicity constraint: Penalize non-monotonic wear trajectories
    3. Irreversibility constraint: dW/dt >= 0
    4. Energy consistency: Force-wear proportionality (Archard)
    5. Uncertainty regularization: NLL with learned variance
    6. Smoothness regularization: Penalize abrupt wear rate changes
    7. ODE consistency: Decoded wear should match ODE-predicted wear
    """

    def __init__(self, lambda_mono=5.0, lambda_irreversible=10.0,
                 lambda_energy=1.0, lambda_smooth=0.5,
                 lambda_ode=2.0, lambda_flute=1.0,
                 lambda_rul=1.0, lambda_uncertainty=0.1):
        super().__init__()

        self.lambda_mono = lambda_mono
        self.lambda_irreversible = lambda_irreversible
        self.lambda_energy = lambda_energy
        self.lambda_smooth = lambda_smooth
        self.lambda_ode = lambda_ode
        self.lambda_flute = lambda_flute
        self.lambda_rul = lambda_rul
        self.lambda_uncertainty = lambda_uncertainty

        self.mse = nn.MSELoss()
        self.huber = nn.SmoothL1Loss()

    def monotonicity_loss(self, wear_pred):
        """Penalize violations of monotonic increase in wear."""
        diff = wear_pred[:, 1:] - wear_pred[:, :-1]
        violations = F.relu(-diff)  # Only penalize decreases
        return torch.mean(violations ** 2)

    def irreversibility_loss(self, wear_ode):
        """Strongly penalize any decrease in ODE wear state."""
        diff = wear_ode[:, 1:] - wear_ode[:, :-1]
        violations = F.relu(-diff)
        return torch.mean(violations ** 2)

    def smoothness_loss(self, wear_pred):
        """Penalize abrupt changes in wear rate (2nd derivative)."""
        first_diff = wear_pred[:, 1:] - wear_pred[:, :-1]
        second_diff = first_diff[:, 1:] - first_diff[:, :-1]
        return torch.mean(second_diff ** 2)

    def ode_consistency_loss(self, wear_pred, wear_ode):
        """Decoded wear should be consistent with ODE-predicted wear."""
        return self.mse(wear_pred, wear_ode)

    def nll_loss(self, pred, target, log_var):
        """Negative log-likelihood with learned variance."""
        precision = torch.exp(-log_var)
        nll = 0.5 * (precision * (pred - target)**2 + log_var)
        return torch.mean(nll)

    def forward(self, outputs, targets):
        """
        Compute total physics-informed loss.

        Args:
            outputs: dict from model forward pass
            targets: dict with 'wear', 'rul', 'flute_wear' tensors
        """
        losses = {}

        # 1. Data fidelity losses
        losses['wear'] = self.huber(outputs['wear_pred'], targets['wear'])
        losses['rul'] = self.huber(outputs['rul_pred'], targets['rul'])
        losses['flute'] = self.huber(outputs['flute_pred'], targets['flute_wear'])

        # 2. Physics constraint losses
        losses['monotonicity'] = self.monotonicity_loss(outputs['wear_pred'])
        losses['irreversibility'] = self.irreversibility_loss(outputs['wear_ode'])
        losses['smoothness'] = self.smoothness_loss(outputs['wear_pred'])

        # 3. ODE consistency
        losses['ode_consistency'] = self.ode_consistency_loss(
            outputs['wear_pred'], outputs['wear_ode'])

        # 4. Uncertainty-aware loss
        losses['nll_wear'] = self.nll_loss(
            outputs['wear_pred'], targets['wear'], outputs['log_var_wear'])
        losses['nll_rul'] = self.nll_loss(
            outputs['rul_pred'], targets['rul'], outputs['log_var_rul'])

        # Total loss
        total = (
            losses['wear']
            + self.lambda_rul * losses['rul']
            + self.lambda_flute * losses['flute']
            + self.lambda_mono * losses['monotonicity']
            + self.lambda_irreversible * losses['irreversibility']
            + self.lambda_smooth * losses['smoothness']
            + self.lambda_ode * losses['ode_consistency']
            + self.lambda_uncertainty * (losses['nll_wear'] + losses['nll_rul'])
        )

        losses['total'] = total
        return losses


criterion = PhysicsInformedLoss()
print("Physics-Informed Loss Function initialized with components:")
for name, val in [
    ("Wear MSE", 1.0), ("RUL", criterion.lambda_rul),
    ("Flute", criterion.lambda_flute), ("Monotonicity", criterion.lambda_mono),
    ("Irreversibility", criterion.lambda_irreversible),
    ("Smoothness", criterion.lambda_smooth), ("ODE Consistency", criterion.lambda_ode),
    ("Uncertainty", criterion.lambda_uncertainty),
]:
    print(f"  λ_{name} = {val}")

# %% [markdown]
# # Cell 9: Training Loop with Physics-Aware Scheduling

# %%
def train_model(model, train_loader, criterion, num_epochs=NUM_EPOCHS,
                lr=LEARNING_RATE, device=DEVICE):
    """
    Training loop with:
    - Cosine annealing warm restarts
    - Gradient clipping
    - Physics constraint curriculum (gradually increase physics penalty)
    - Best model checkpointing
    """

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr,
                                   weight_decay=WEIGHT_DECAY)
    scheduler = CosineAnnealingWarmRestarts(optimizer, T_0=50, T_mult=2, eta_min=1e-6)

    history = {
        'total': [], 'wear': [], 'rul': [], 'monotonicity': [],
        'irreversibility': [], 'ode_consistency': [], 'lr': [],
    }

    best_loss = float('inf')
    best_state = None
    patience_counter = 0
    patience = 40

    print(f"\n{'='*60}")
    print(f"TRAINING STARTED")
    print(f"{'='*60}")
    print(f"Epochs: {num_epochs}, LR: {lr}, Batch: {BATCH_SIZE}")
    print(f"{'='*60}\n")

    for epoch in range(1, num_epochs + 1):
        model.train()
        epoch_losses = {k: 0.0 for k in history.keys() if k != 'lr'}
        n_batches = 0

        # Physics constraint curriculum: ramp up over first 50 epochs
        curriculum_weight = min(1.0, epoch / 50.0)
        criterion.lambda_mono = 5.0 * curriculum_weight
        criterion.lambda_irreversible = 10.0 * curriculum_weight

        for batch in train_loader:
            feat_seq, wear_seq, rul_seq, flute_seq, time_seq = batch
            feat_seq = feat_seq.to(device)
            wear_seq = wear_seq.to(device)
            rul_seq = rul_seq.to(device)
            flute_seq = flute_seq.to(device)
            time_seq = time_seq.to(device)

            # Forward pass
            outputs = model(feat_seq, time_seq)

            targets = {
                'wear': wear_seq,
                'rul': rul_seq,
                'flute_wear': flute_seq,
            }

            losses = criterion(outputs, targets)

            # Backward pass
            optimizer.zero_grad()
            losses['total'].backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()

            for k in epoch_losses:
                if k in losses:
                    epoch_losses[k] += losses[k].item()
            n_batches += 1

        scheduler.step()

        # Average losses
        for k in epoch_losses:
            epoch_losses[k] /= max(n_batches, 1)
            history[k].append(epoch_losses[k])
        history['lr'].append(optimizer.param_groups[0]['lr'])

        # Best model tracking
        if epoch_losses['total'] < best_loss:
            best_loss = epoch_losses['total']
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1

        # Print progress
        if epoch % 10 == 0 or epoch == 1:
            print(f"Epoch {epoch:03d}/{num_epochs} | "
                  f"Total: {epoch_losses['total']:.6f} | "
                  f"Wear: {epoch_losses['wear']:.6f} | "
                  f"RUL: {epoch_losses['rul']:.6f} | "
                  f"Mono: {epoch_losses['monotonicity']:.6f} | "
                  f"ODE: {epoch_losses['ode_consistency']:.6f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.2e}")

        # Early stopping
        if patience_counter >= patience:
            print(f"\nEarly stopping at epoch {epoch} (patience={patience})")
            break

    # Restore best model
    if best_state is not None:
        model.load_state_dict(best_state)
        print(f"\nRestored best model (loss: {best_loss:.6f})")

    return history


# ===== TRAIN =====
history = train_model(model, train_loader, criterion)

# %% [markdown]
# # Cell 10: Training Curves Visualization

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Total loss
axes[0, 0].plot(history['total'], 'b-', linewidth=1.5)
axes[0, 0].set_title('Total Loss')
axes[0, 0].set_xlabel('Epoch')
axes[0, 0].set_ylabel('Loss')
axes[0, 0].set_yscale('log')
axes[0, 0].grid(True, alpha=0.3)

# Wear loss
axes[0, 1].plot(history['wear'], 'r-', linewidth=1.5)
axes[0, 1].set_title('Wear Prediction Loss')
axes[0, 1].set_xlabel('Epoch')
axes[0, 1].set_ylabel('Loss')
axes[0, 1].set_yscale('log')
axes[0, 1].grid(True, alpha=0.3)

# RUL loss
axes[0, 2].plot(history['rul'], 'g-', linewidth=1.5)
axes[0, 2].set_title('RUL Prediction Loss')
axes[0, 2].set_xlabel('Epoch')
axes[0, 2].set_ylabel('Loss')
axes[0, 2].set_yscale('log')
axes[0, 2].grid(True, alpha=0.3)

# Monotonicity loss
axes[1, 0].plot(history['monotonicity'], 'm-', linewidth=1.5)
axes[1, 0].set_title('Monotonicity Violation Loss')
axes[1, 0].set_xlabel('Epoch')
axes[1, 0].set_ylabel('Loss')
axes[1, 0].set_yscale('log')
axes[1, 0].grid(True, alpha=0.3)

# ODE consistency loss
axes[1, 1].plot(history['ode_consistency'], 'c-', linewidth=1.5)
axes[1, 1].set_title('ODE Consistency Loss')
axes[1, 1].set_xlabel('Epoch')
axes[1, 1].set_ylabel('Loss')
axes[1, 1].set_yscale('log')
axes[1, 1].grid(True, alpha=0.3)

# Learning rate
axes[1, 2].plot(history['lr'], 'k-', linewidth=1.5)
axes[1, 2].set_title('Learning Rate Schedule')
axes[1, 2].set_xlabel('Epoch')
axes[1, 2].set_ylabel('LR')
axes[1, 2].set_yscale('log')
axes[1, 2].grid(True, alpha=0.3)

plt.suptitle('PINN-NDE Training Dynamics', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('training_curves.png', dpi=300, bbox_inches='tight')
plt.show()
print("Training curves saved.")

# %% [markdown]
# # Cell 11: Full-Trajectory Inference Function

# %%
@torch.no_grad()
def predict_full_trajectory(model, features, seq_len=SEQ_LEN, device=DEVICE):
    """
    Run inference on full cutter trajectory using sliding window.
    Returns predictions for every cut by averaging overlapping windows.

    Args:
        features: (N_cuts, feature_dim) normalized features
    Returns:
        dict with wear, rul, flute, uncertainty predictions for each cut
    """
    model.eval()
    n_cuts = len(features)

    # Accumulators for averaging overlapping predictions
    wear_accum = np.zeros(n_cuts)
    rul_accum = np.zeros(n_cuts)
    flute_accum = np.zeros((n_cuts, NUM_FLUTES))
    var_wear_accum = np.zeros(n_cuts)
    var_rul_accum = np.zeros(n_cuts)
    phase_accum = np.zeros((n_cuts, 3))
    temp_accum = np.zeros(n_cuts)
    stress_accum = np.zeros(n_cuts)
    count = np.zeros(n_cuts)

    for i in range(n_cuts - seq_len):
        feat_seq = torch.FloatTensor(features[i:i+seq_len]).unsqueeze(0).to(device)
        time_seq = torch.FloatTensor(
            np.arange(i, i+seq_len, dtype=np.float32) / n_cuts
        ).unsqueeze(0).to(device)

        outputs = model(feat_seq, time_seq)

        wear = outputs['wear_pred'][0].cpu().numpy()
        rul = outputs['rul_pred'][0].cpu().numpy()
        flute = outputs['flute_pred'][0].cpu().numpy()
        var_w = torch.exp(outputs['log_var_wear'][0]).cpu().numpy()
        var_r = torch.exp(outputs['log_var_rul'][0]).cpu().numpy()
        phase = F.softmax(outputs['phase_logits'][0], dim=-1).cpu().numpy()
        temp = outputs['latent_temperature'][0].cpu().numpy()
        stress = outputs['latent_stress'][0].cpu().numpy()

        for j in range(seq_len):
            idx = i + j
            wear_accum[idx] += wear[j]
            rul_accum[idx] += rul[j]
            flute_accum[idx] += flute[j]
            var_wear_accum[idx] += var_w[j]
            var_rul_accum[idx] += var_r[j]
            phase_accum[idx] += phase[j]
            temp_accum[idx] += temp[j]
            stress_accum[idx] += stress[j]
            count[idx] += 1

    # Avoid division by zero
    count = np.maximum(count, 1)

    results = {
        'wear': wear_accum / count,
        'rul': rul_accum / count,
        'flute': flute_accum / count[:, None],
        'wear_std': np.sqrt(var_wear_accum / count),
        'rul_std': np.sqrt(var_rul_accum / count),
        'phase_probs': phase_accum / count[:, None],
        'latent_temperature': temp_accum / count,
        'latent_stress': stress_accum / count,
    }

    return results


# Run inference on training cutters
print("Running inference on training cutters...")
train_predictions = {}
for cid in TRAIN_CUTTERS:
    train_predictions[cid] = predict_full_trajectory(model, train_features[cid])
    print(f"  {cid}: Predicted {len(train_predictions[cid]['wear'])} cuts")

# Run inference on test cutters
print("\nRunning inference on test cutters...")
test_predictions = {}
for cid in TEST_CUTTERS:
    test_predictions[cid] = predict_full_trajectory(model, test_features[cid])
    print(f"  {cid}: Predicted {len(test_predictions[cid]['wear'])} cuts")

# %% [markdown]
# # Cell 12: Quantitative Evaluation Metrics

# %%
def compute_metrics(y_true, y_pred):
    """Compute comprehensive regression metrics."""
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / (y_true + 1e-10))) * 100

    # Correlation metrics
    pearson_r, _ = pearsonr(y_true, y_pred)
    spearman_r, _ = spearmanr(y_true, y_pred)

    # Monotonicity score
    diffs = np.diff(y_pred)
    mono_score = np.sum(diffs >= 0) / len(diffs) * 100

    return {
        'RMSE': rmse, 'MAE': mae, 'R²': r2, 'MAPE': mape,
        'Pearson': pearson_r, 'Spearman': spearman_r,
        'Monotonicity%': mono_score
    }


print("=" * 80)
print("QUANTITATIVE EVALUATION - TRAINING CUTTERS (Wear Prediction)")
print("=" * 80)

train_metrics_table = []
for cid in TRAIN_CUTTERS:
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    # Align lengths
    min_len = min(len(y_true), len(y_pred))
    metrics = compute_metrics(y_true[:min_len], y_pred[:min_len])
    metrics['Cutter'] = cid.upper()
    train_metrics_table.append(metrics)
    print(f"\n{cid.upper()}:")
    for k, v in metrics.items():
        if k != 'Cutter':
            print(f"  {k:15s}: {v:.4f}")

print("\n" + "=" * 80)
print("QUANTITATIVE EVALUATION - TEST CUTTERS (Wear Prediction)")
print("=" * 80)
print("(Note: Test cutters have no wear labels; evaluating trajectory plausibility)")
for cid in TEST_CUTTERS:
    pred = test_predictions[cid]['wear']
    diffs = np.diff(pred)
    mono = np.sum(diffs >= 0) / len(diffs) * 100
    print(f"\n{cid.upper()}:")
    print(f"  Wear range:      [{pred.min():.4f}, {pred.max():.4f}]")
    print(f"  Monotonicity:    {mono:.1f}%")
    print(f"  Mean uncertainty: {test_predictions[cid]['wear_std'].mean():.4f}")

# Summary table
print("\n" + "=" * 80)
print("SUMMARY TABLE (Training Cutters)")
print("=" * 80)
df_metrics = pd.DataFrame(train_metrics_table)
df_metrics = df_metrics.set_index('Cutter')
print(df_metrics.to_string(float_format='%.4f'))

# %% [markdown]
# # Cell 13: Publication-Quality Wear Prediction Plots (Training)

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

colors = {'c1': '#2196F3', 'c4': '#E91E63', 'c6': '#4CAF50'}

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    y_std = train_predictions[cid]['wear_std']
    min_len = min(len(y_true), len(y_pred))

    cuts = np.arange(1, min_len + 1)

    ax.plot(cuts, y_true[:min_len], 'k-', linewidth=2, label='Ground Truth', zorder=3)
    ax.plot(cuts, y_pred[:min_len], '-', color=colors[cid], linewidth=1.5,
            label='PINN-NDE Prediction', zorder=2)
    ax.fill_between(cuts,
                    y_pred[:min_len] - 2*y_std[:min_len],
                    y_pred[:min_len] + 2*y_std[:min_len],
                    alpha=0.2, color=colors[cid], label='95% CI', zorder=1)

    metrics = compute_metrics(y_true[:min_len], y_pred[:min_len])
    ax.text(0.05, 0.95,
            f"RMSE = {metrics['RMSE']:.4f}\n"
            f"R² = {metrics['R²']:.4f}\n"
            f"Mono = {metrics['Monotonicity%']:.1f}%",
            transform=ax.transAxes, fontsize=10,
            verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    ax.set_title(f'Cutter {cid.upper()} (Training)', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(loc='lower right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Tool Wear Prediction with Uncertainty Quantification',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('wear_prediction_training.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 14: Test Cutter Predictions (Generalization)

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

test_colors = {'c2': '#FF9800', 'c3': '#9C27B0', 'c5': '#00BCD4'}

for idx, cid in enumerate(TEST_CUTTERS):
    ax = axes[idx]
    y_pred = test_predictions[cid]['wear']
    y_std = test_predictions[cid]['wear_std']
    cuts = np.arange(1, len(y_pred) + 1)

    ax.plot(cuts, y_pred, '-', color=test_colors[cid], linewidth=1.5,
            label='PINN-NDE Prediction')
    ax.fill_between(cuts,
                    y_pred - 2*y_std,
                    y_pred + 2*y_std,
                    alpha=0.2, color=test_colors[cid], label='95% CI')

    diffs = np.diff(y_pred)
    mono = np.sum(diffs >= 0) / len(diffs) * 100
    ax.text(0.05, 0.95,
            f"Mono = {mono:.1f}%\nσ_mean = {y_std.mean():.4f}",
            transform=ax.transAxes, fontsize=10,
            verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    ax.set_title(f'Cutter {cid.upper()} (Test)', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(loc='lower right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Generalization to Unseen Test Cutters',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('wear_prediction_test.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 15: Flute-Level Wear Decomposition Plot

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
flute_colors = ['#F44336', '#2196F3', '#4CAF50']

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    w_data = train_wear[cid]
    pred_flute = train_predictions[cid]['flute']
    min_len = min(len(w_data), len(pred_flute))
    cuts = np.arange(1, min_len + 1)

    for f in range(3):
        # Ground truth
        gt = wear_scaler.transform(
            w_data[f'flute_{f+1}'].values[:min_len].reshape(-1, 1)).flatten()
        ax.plot(cuts, gt, '--', color=flute_colors[f], linewidth=1.5, alpha=0.7,
                label=f'Flute {f+1} GT')
        # Prediction
        ax.plot(cuts, pred_flute[:min_len, f], '-', color=flute_colors[f],
                linewidth=1.5, label=f'Flute {f+1} Pred')

    ax.set_title(f'Cutter {cid.upper()} - Flute Decomposition', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(loc='upper left', fontsize=8, ncol=2)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Flute-Level Wear Decomposition',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('flute_wear_decomposition.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 16: RUL Prediction Analysis

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    y_true_rul = train_rul[cid]
    y_pred_rul = train_predictions[cid]['rul']
    y_std_rul = train_predictions[cid]['rul_std']
    min_len = min(len(y_true_rul), len(y_pred_rul))
    cuts = np.arange(1, min_len + 1)

    ax.plot(cuts, y_true_rul[:min_len], 'k-', linewidth=2, label='Ground Truth RUL')
    ax.plot(cuts, y_pred_rul[:min_len], '-', color=colors[cid], linewidth=1.5,
            label='PINN-NDE Prediction')
    ax.fill_between(cuts,
                    y_pred_rul[:min_len] - 2*y_std_rul[:min_len],
                    y_pred_rul[:min_len] + 2*y_std_rul[:min_len],
                    alpha=0.2, color=colors[cid], label='95% CI')

    metrics = compute_metrics(y_true_rul[:min_len], y_pred_rul[:min_len])
    ax.text(0.60, 0.95,
            f"RMSE = {metrics['RMSE']:.4f}\nR² = {metrics['R²']:.4f}",
            transform=ax.transAxes, fontsize=10,
            verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    ax.set_title(f'Cutter {cid.upper()} - RUL', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized RUL')
    ax.legend(loc='center right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Remaining Useful Life Prediction',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('rul_prediction.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 17: Wear Phase Transition Analysis

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
phase_labels = ['Initial\n(Run-in)', 'Steady\n(Normal)', 'Accelerated\n(Failure)']
phase_colors = ['#4CAF50', '#2196F3', '#F44336']

for idx, cid in enumerate(TRAIN_CUTTERS):
    # Phase probabilities
    ax_top = axes[0, idx]
    phase_probs = train_predictions[cid]['phase_probs']
    min_len = min(len(phase_probs), NUM_CUTS)
    cuts = np.arange(1, min_len + 1)

    for p in range(3):
        ax_top.fill_between(cuts, 0, phase_probs[:min_len, p],
                           alpha=0.3, color=phase_colors[p])
        ax_top.plot(cuts, phase_probs[:min_len, p], '-',
                   color=phase_colors[p], linewidth=1.5, label=phase_labels[p])

    ax_top.set_title(f'Cutter {cid.upper()} - Phase Probabilities', fontweight='bold')
    ax_top.set_xlabel('Cut Number')
    ax_top.set_ylabel('Probability')
    ax_top.legend(loc='center right', fontsize=8)
    ax_top.set_ylim(0, 1.05)
    ax_top.grid(True, alpha=0.3)

    # Dominant phase overlay on wear
    ax_bot = axes[1, idx]
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    min_len2 = min(len(y_true), len(y_pred))
    cuts2 = np.arange(1, min_len2 + 1)

    dominant_phase = np.argmax(phase_probs[:min_len2], axis=1)

    ax_bot.plot(cuts2, y_true[:min_len2], 'k-', linewidth=2, label='Ground Truth')
    for p in range(3):
        mask = dominant_phase == p
        if np.any(mask):
            ax_bot.scatter(cuts2[mask], y_pred[:min_len2][mask], c=phase_colors[p],
                          s=10, alpha=0.7, label=phase_labels[p].replace('\n', ' '))

    ax_bot.set_title(f'Cutter {cid.upper()} - Phase-Colored Wear', fontweight='bold')
    ax_bot.set_xlabel('Cut Number')
    ax_bot.set_ylabel('Normalized Wear')
    ax_bot.legend(loc='upper left', fontsize=8)
    ax_bot.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Wear Phase Transition Detection',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('phase_transition_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 18: Latent Physics Inference (Temperature & Stress Proxies)

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for idx, cid in enumerate(TRAIN_CUTTERS):
    temp = train_predictions[cid]['latent_temperature']
    stress = train_predictions[cid]['latent_stress']
    wear = train_wear_normalized[cid]
    min_len = min(len(temp), len(wear))
    cuts = np.arange(1, min_len + 1)

    # Temperature proxy
    ax_temp = axes[0, idx]
    ax_temp.plot(cuts, temp[:min_len], '-', color='#FF5722', linewidth=1.5)
    ax_temp.set_title(f'{cid.upper()} - Inferred Temperature Proxy', fontweight='bold')
    ax_temp.set_xlabel('Cut Number')
    ax_temp.set_ylabel('Temperature (a.u.)')
    ax_temp.grid(True, alpha=0.3)

    # Add correlation with wear
    corr, _ = pearsonr(temp[:min_len], wear[:min_len])
    ax_temp.text(0.05, 0.95, f'ρ(T, W) = {corr:.3f}',
                transform=ax_temp.transAxes, fontsize=10,
                verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    # Stress proxy
    ax_stress = axes[1, idx]
    ax_stress.plot(cuts, stress[:min_len], '-', color='#3F51B5', linewidth=1.5)
    ax_stress.set_title(f'{cid.upper()} - Inferred Stress Proxy', fontweight='bold')
    ax_stress.set_xlabel('Cut Number')
    ax_stress.set_ylabel('Contact Stress (a.u.)')
    ax_stress.grid(True, alpha=0.3)

    corr_s, _ = pearsonr(stress[:min_len], wear[:min_len])
    ax_stress.text(0.05, 0.95, f'ρ(σ, W) = {corr_s:.3f}',
                  transform=ax_stress.transAxes, fontsize=10,
                  verticalalignment='top',
                  bbox=dict(boxstyle='round', facecolor='lightcyan', alpha=0.8))

plt.suptitle('PINN-NDE: Latent Physics Variable Inference\n'
             '(Temperature & Contact Stress from Sensor Signals)',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('latent_physics_inference.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 19: Physics Parameter Analysis (Discovered Wear Law)

# %%
print("=" * 60)
print("DISCOVERED PHYSICS PARAMETERS (Archard-Usui Hybrid)")
print("=" * 60)
print("\nLearned wear rate equation:")
print("  dW/dt = a · exp(b · W) · (1 + c · t) + f_neural(z, W, t)")
print()

with torch.no_grad():
    sp = nn.Softplus(beta=5)
    a = sp(model.ode_func.physics_a).item()
    b = sp(model.ode_func.physics_b).item()
    c = sp(model.ode_func.physics_c).item()

print(f"  a (base wear rate):         {a:.6f}")
print(f"  b (wear acceleration coeff): {b:.6f}")
print(f"  c (time-dependent factor):   {c:.6f}")
print()
print("Physical interpretation:")
print(f"  - Base rate a = {a:.4f}: Controls initial wear speed")
print(f"  - Acceleration b = {b:.4f}: Exponential wear growth "
      f"(higher = faster transition to accelerated phase)")
print(f"  - Time factor c = {c:.4f}: Progressive hardening/softening effect")

# Visualize the physics component of wear rate
W_range = torch.linspace(0, 1, 100).to(DEVICE)
t_values = [0.0, 0.25, 0.5, 0.75, 1.0]

fig, ax = plt.subplots(figsize=(10, 6))
for t_val in t_values:
    t_tensor = torch.tensor(t_val).to(DEVICE) * torch.ones_like(W_range.unsqueeze(-1))
    with torch.no_grad():
        rate = model.ode_func.physics_rate(W_range.unsqueeze(-1), t_tensor)
    ax.plot(W_range.cpu().numpy(), rate.cpu().numpy().flatten(),
            linewidth=2, label=f't = {t_val:.2f}')

ax.set_xlabel('Normalized Wear (W)', fontsize=13)
ax.set_ylabel('Physics Wear Rate dW/dt', fontsize=13)
ax.set_title('Discovered Physics-Based Wear Rate Function\n'
             r'$\frac{dW}{dt}_{phys} = a \cdot e^{bW} \cdot (1 + ct)$',
             fontsize=14, fontweight='bold')
ax.legend(title='Normalized Time', fontsize=10)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('discovered_wear_law.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 20: Ablation Study - Physics Constraints Contribution

# %%
def train_ablation_model(model_class, train_loader, config_name,
                         lambda_mono=0.0, lambda_irreversible=0.0,
                         lambda_ode=0.0, num_epochs=150, device=DEVICE):
    """Train a model variant for ablation study."""
    print(f"\n--- Ablation: {config_name} ---")

    ablation_model = model_class().to(device)
    ablation_criterion = PhysicsInformedLoss(
        lambda_mono=lambda_mono,
        lambda_irreversible=lambda_irreversible,
        lambda_ode=lambda_ode,
    )

    optimizer = torch.optim.AdamW(ablation_model.parameters(), lr=LEARNING_RATE,
                                   weight_decay=WEIGHT_DECAY)
    scheduler = CosineAnnealingWarmRestarts(optimizer, T_0=50, T_mult=2, eta_min=1e-6)

    best_loss = float('inf')
    best_state = None

    for epoch in range(1, num_epochs + 1):
        ablation_model.train()
        total_loss = 0
        n_batches = 0

        for batch in train_loader:
            feat_seq, wear_seq, rul_seq, flute_seq, time_seq = batch
            feat_seq = feat_seq.to(device)
            wear_seq = wear_seq.to(device)
            rul_seq = rul_seq.to(device)
            flute_seq = flute_seq.to(device)
            time_seq = time_seq.to(device)

            outputs = ablation_model(feat_seq, time_seq)
            targets = {'wear': wear_seq, 'rul': rul_seq, 'flute_wear': flute_seq}
            losses = ablation_criterion(outputs, targets)

            optimizer.zero_grad()
            losses['total'].backward()
            torch.nn.utils.clip_grad_norm_(ablation_model.parameters(), max_norm=1.0)
            optimizer.step()

            total_loss += losses['total'].item()
            n_batches += 1

        scheduler.step()
        avg_loss = total_loss / max(n_batches, 1)

        if avg_loss < best_loss:
            best_loss = avg_loss
            best_state = {k: v.cpu().clone() for k, v in ablation_model.state_dict().items()}

        if epoch % 50 == 0:
            print(f"  Epoch {epoch}: Loss = {avg_loss:.6f}")

    if best_state:
        ablation_model.load_state_dict(best_state)

    return ablation_model


print("=" * 60)
print("ABLATION STUDY: Physics Constraints Contribution")
print("=" * 60)

# Ablation configurations
ablation_configs = {
    'Full PINN-NDE': {'lambda_mono': 5.0, 'lambda_irreversible': 10.0, 'lambda_ode': 2.0},
    'No Monotonicity': {'lambda_mono': 0.0, 'lambda_irreversible': 10.0, 'lambda_ode': 2.0},
    'No Irreversibility': {'lambda_mono': 5.0, 'lambda_irreversible': 0.0, 'lambda_ode': 2.0},
    'No ODE Consistency': {'lambda_mono': 5.0, 'lambda_irreversible': 10.0, 'lambda_ode': 0.0},
    'Purely Data-Driven': {'lambda_mono': 0.0, 'lambda_irreversible': 0.0, 'lambda_ode': 0.0},
}

ablation_results = {}
for config_name, config in ablation_configs.items():
    if config_name == 'Full PINN-NDE':
        # Use already trained model
        abl_model = model
    else:
        abl_model = train_ablation_model(
            PhysicsInformedNeuralWearModel, train_loader,
            config_name, num_epochs=150, **config
        )

    # Evaluate on c1 (representative training cutter)
    preds = predict_full_trajectory(abl_model, train_features['c1'])
    y_true = train_wear_normalized['c1']
    min_len = min(len(y_true), len(preds['wear']))
    metrics = compute_metrics(y_true[:min_len], preds['wear'][:min_len])
    ablation_results[config_name] = metrics
    print(f"\n{config_name}: RMSE={metrics['RMSE']:.4f}, "
          f"R²={metrics['R²']:.4f}, Mono={metrics['Monotonicity%']:.1f}%")

# Ablation summary table
print("\n" + "=" * 80)
print("ABLATION STUDY SUMMARY (Cutter C1)")
print("=" * 80)
df_ablation = pd.DataFrame(ablation_results).T
print(df_ablation[['RMSE', 'MAE', 'R²', 'Monotonicity%']].to_string(float_format='%.4f'))

# Plot ablation comparison
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
abl_names = list(ablation_results.keys())
abl_rmse = [ablation_results[n]['RMSE'] for n in abl_names]
abl_r2 = [ablation_results[n]['R²'] for n in abl_names]
abl_mono = [ablation_results[n]['Monotonicity%'] for n in abl_names]

bar_colors = ['#4CAF50', '#FF9800', '#F44336', '#9C27B0', '#607D8B']

axes[0].barh(abl_names, abl_rmse, color=bar_colors)
axes[0].set_xlabel('RMSE ↓')
axes[0].set_title('RMSE Comparison', fontweight='bold')
axes[0].invert_xaxis()

axes[1].barh(abl_names, abl_r2, color=bar_colors)
axes[1].set_xlabel('R² ↑')
axes[1].set_title('R² Comparison', fontweight='bold')

axes[2].barh(abl_names, abl_mono, color=bar_colors)
axes[2].set_xlabel('Monotonicity % ↑')
axes[2].set_title('Monotonicity Comparison', fontweight='bold')

plt.suptitle('Ablation Study: Contribution of Physics Constraints',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('ablation_study.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 21: Comparison with Baseline Methods

# %%
# ========= BASELINE 1: Linear Regression =========
from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor

print("=" * 60)
print("BASELINE COMPARISONS")
print("=" * 60)

baseline_results = {}

# Prepare flat data for baselines
X_train_flat = np.vstack([train_features[cid] for cid in TRAIN_CUTTERS])
y_train_flat = np.concatenate([train_wear_normalized[cid] for cid in TRAIN_CUTTERS])

# Ridge Regression
print("\n--- Ridge Regression ---")
ridge = Ridge(alpha=1.0)
ridge.fit(X_train_flat, y_train_flat)
baseline_preds = {}
for cid in TRAIN_CUTTERS:
    pred = ridge.predict(train_features[cid])
    pred = np.clip(pred, 0, 1)
    min_len = min(len(train_wear_normalized[cid]), len(pred))
    metrics = compute_metrics(train_wear_normalized[cid][:min_len], pred[:min_len])
    baseline_preds[f'Ridge_{cid}'] = pred
    print(f"  {cid}: RMSE={metrics['RMSE']:.4f}, R²={metrics['R²']:.4f}")
baseline_results['Ridge Regression'] = metrics

# Gradient Boosting
print("\n--- Gradient Boosting ---")
gb = GradientBoostingRegressor(n_estimators=200, max_depth=5, learning_rate=0.05,
                                random_state=SEED)
gb.fit(X_train_flat, y_train_flat)
for cid in TRAIN_CUTTERS:
    pred = gb.predict(train_features[cid])
    pred = np.clip(pred, 0, 1)
    min_len = min(len(train_wear_normalized[cid]), len(pred))
    metrics = compute_metrics(train_wear_normalized[cid][:min_len], pred[:min_len])
    baseline_preds[f'GB_{cid}'] = pred
    print(f"  {cid}: RMSE={metrics['RMSE']:.4f}, R²={metrics['R²']:.4f}")
baseline_results['Gradient Boosting'] = metrics

# Simple LSTM baseline
print("\n--- LSTM Baseline ---")

class SimpleLSTM(nn.Module):
    def __init__(self, input_dim=FEATURE_DIM, hidden_dim=64):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers=2,
                           batch_first=True, dropout=0.1)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x, _=None):
        out, _ = self.lstm(x)
        pred = self.fc(out).squeeze(-1)
        return {'wear_pred': pred, 'rul_pred': pred,
                'flute_pred': pred.unsqueeze(-1).expand(-1, -1, 3),
                'wear_ode': pred,
                'log_var_wear': torch.zeros_like(pred),
                'log_var_rul': torch.zeros_like(pred),
                'phase_logits': torch.zeros(pred.shape[0], pred.shape[1], 3, device=pred.device),
                'latent_temperature': torch.zeros_like(pred),
                'latent_stress': torch.zeros_like(pred),
                'latent_z': torch.zeros(pred.shape[0], pred.shape[1], LATENT_DIM, device=pred.device)}

lstm_model = SimpleLSTM().to(DEVICE)
lstm_criterion = PhysicsInformedLoss(lambda_mono=0, lambda_irreversible=0,
                                      lambda_ode=0, lambda_uncertainty=0)
lstm_opt = torch.optim.Adam(lstm_model.parameters(), lr=1e-3)

for epoch in range(1, 151):
    lstm_model.train()
    for batch in train_loader:
        feat_seq, wear_seq, rul_seq, flute_seq, time_seq = batch
        feat_seq, wear_seq = feat_seq.to(DEVICE), wear_seq.to(DEVICE)
        rul_seq, flute_seq = rul_seq.to(DEVICE), flute_seq.to(DEVICE)
        time_seq = time_seq.to(DEVICE)

        outputs = lstm_model(feat_seq, time_seq)
        targets = {'wear': wear_seq, 'rul': rul_seq, 'flute_wear': flute_seq}
        losses = lstm_criterion(outputs, targets)

        lstm_opt.zero_grad()
        losses['total'].backward()
        lstm_opt.step()

    if epoch % 50 == 0:
        print(f"  Epoch {epoch}")

for cid in TRAIN_CUTTERS:
    lstm_preds = predict_full_trajectory(lstm_model, train_features[cid])
    y_true = train_wear_normalized[cid]
    min_len = min(len(y_true), len(lstm_preds['wear']))
    metrics = compute_metrics(y_true[:min_len], lstm_preds['wear'][:min_len])
    print(f"  {cid}: RMSE={metrics['RMSE']:.4f}, R²={metrics['R²']:.4f}, "
          f"Mono={metrics['Monotonicity%']:.1f}%")
baseline_results['LSTM'] = metrics

# CNN baseline
print("\n--- 1D-CNN Baseline ---")

class SimpleCNN(nn.Module):
    def __init__(self, input_dim=FEATURE_DIM, seq_len=SEQ_LEN):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv1d(input_dim, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv1d(64, 32, kernel_size=3, padding=1),
            nn.ReLU(),
        )
        self.fc = nn.Sequential(
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, x, _=None):
        # x: (B, T, F) -> (B, F, T)
        out = self.conv(x.transpose(1, 2))  # (B, 32, T)
        out = out.transpose(1, 2)  # (B, T, 32)
        pred = self.fc(out).squeeze(-1)
        return {'wear_pred': pred, 'rul_pred': pred,
                'flute_pred': pred.unsqueeze(-1).expand(-1, -1, 3),
                'wear_ode': pred,
                'log_var_wear': torch.zeros_like(pred),
                'log_var_rul': torch.zeros_like(pred),
                'phase_logits': torch.zeros(pred.shape[0], pred.shape[1], 3, device=pred.device),
                'latent_temperature': torch.zeros_like(pred),
                'latent_stress': torch.zeros_like(pred),
                'latent_z': torch.zeros(pred.shape[0], pred.shape[1], LATENT_DIM, device=pred.device)}

cnn_model = SimpleCNN().to(DEVICE)
cnn_opt = torch.optim.Adam(cnn_model.parameters(), lr=1e-3)

for epoch in range(1, 151):
    cnn_model.train()
    for batch in train_loader:
        feat_seq, wear_seq, rul_seq, flute_seq, time_seq = batch
        feat_seq, wear_seq = feat_seq.to(DEVICE), wear_seq.to(DEVICE)
        rul_seq, flute_seq = rul_seq.to(DEVICE), flute_seq.to(DEVICE)
        time_seq = time_seq.to(DEVICE)

        outputs = cnn_model(feat_seq, time_seq)
        targets = {'wear': wear_seq, 'rul': rul_seq, 'flute_wear': flute_seq}
        losses = lstm_criterion(outputs, targets)

        cnn_opt.zero_grad()
        losses['total'].backward()
        cnn_opt.step()

    if epoch % 50 == 0:
        print(f"  Epoch {epoch}")

for cid in TRAIN_CUTTERS:
    cnn_preds = predict_full_trajectory(cnn_model, train_features[cid])
    y_true = train_wear_normalized[cid]
    min_len = min(len(y_true), len(cnn_preds['wear']))
    metrics = compute_metrics(y_true[:min_len], cnn_preds['wear'][:min_len])
    print(f"  {cid}: RMSE={metrics['RMSE']:.4f}, R²={metrics['R²']:.4f}, "
          f"Mono={metrics['Monotonicity%']:.1f}%")
baseline_results['1D-CNN'] = metrics

# Add our full model results
for cid in TRAIN_CUTTERS:
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    min_len = min(len(y_true), len(y_pred))
    metrics = compute_metrics(y_true[:min_len], y_pred[:min_len])
baseline_results['PINN-NDE (Ours)'] = metrics

# Final comparison table
print("\n" + "=" * 80)
print("COMPARISON TABLE (Last Training Cutter)")
print("=" * 80)
df_comparison = pd.DataFrame(baseline_results).T
print(df_comparison[['RMSE', 'MAE', 'R²', 'MAPE', 'Monotonicity%']].to_string(
    float_format='%.4f'))

# %% [markdown]
# # Cell 22: Comprehensive Comparison Visualization

# %%
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

methods = list(baseline_results.keys())
rmse_vals = [baseline_results[m]['RMSE'] for m in methods]
r2_vals = [baseline_results[m]['R²'] for m in methods]

method_colors = ['#607D8B', '#FF9800', '#9C27B0', '#2196F3', '#4CAF50']

# RMSE comparison
bars1 = axes[0].barh(methods, rmse_vals, color=method_colors, edgecolor='black', linewidth=0.5)
axes[0].set_xlabel('RMSE (lower is better)', fontsize=12)
axes[0].set_title('RMSE Comparison', fontweight='bold')
for bar, val in zip(bars1, rmse_vals):
    axes[0].text(bar.get_width() + 0.001, bar.get_y() + bar.get_height()/2,
                f'{val:.4f}', va='center', fontsize=10)
axes[0].grid(True, alpha=0.3, axis='x')

# R² comparison
bars2 = axes[1].barh(methods, r2_vals, color=method_colors, edgecolor='black', linewidth=0.5)
axes[1].set_xlabel('R² (higher is better)', fontsize=12)
axes[1].set_title('R² Score Comparison', fontweight='bold')
for bar, val in zip(bars2, r2_vals):
    axes[1].text(bar.get_width() + 0.005, bar.get_y() + bar.get_height()/2,
                f'{val:.4f}', va='center', fontsize=10)
axes[1].grid(True, alpha=0.3, axis='x')

plt.suptitle('PINN-NDE vs. Baseline Methods',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('baseline_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 23: Uncertainty Calibration Analysis

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    y_std = train_predictions[cid]['wear_std']
    min_len = min(len(y_true), len(y_pred))

    errors = np.abs(y_true[:min_len] - y_pred[:min_len])
    stds = y_std[:min_len]

    # Check calibration: what % of errors fall within k*sigma
    k_values = np.linspace(0.1, 3.0, 30)
    coverage = []
    expected = []
    for k in k_values:
        cov = np.mean(errors <= k * stds) * 100
        exp = (2 * (1 - 0.5 * (1 + np.math.erf(-k/np.sqrt(2))))) * 100  # Rough approx
        # More precise: use scipy
        from scipy.stats import norm
        exp_precise = (norm.cdf(k) - norm.cdf(-k)) * 100
        coverage.append(cov)
        expected.append(exp_precise)

    ax.plot(expected, coverage, 'o-', color=colors[cid], markersize=4, linewidth=1.5,
            label='Observed')
    ax.plot([0, 100], [0, 100], 'k--', linewidth=1, label='Ideal Calibration')

    ax.set_xlabel('Expected Coverage (%)')
    ax.set_ylabel('Observed Coverage (%)')
    ax.set_title(f'{cid.upper()} - Uncertainty Calibration', fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)
    ax.set_aspect('equal')

plt.suptitle('PINN-NDE: Uncertainty Calibration Analysis',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('uncertainty_calibration.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 24: Latent Space Visualization (t-SNE)

# %%
from sklearn.manifold import TSNE

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]

    # Get latent representations
    model.eval()
    all_z = []
    with torch.no_grad():
        for i in range(0, NUM_CUTS - SEQ_LEN, SEQ_LEN):
            feat = torch.FloatTensor(train_features[cid][i:i+SEQ_LEN]).unsqueeze(0).to(DEVICE)
            time_t = torch.FloatTensor(
                np.arange(i, i+SEQ_LEN, dtype=np.float32) / NUM_CUTS
            ).unsqueeze(0).to(DEVICE)
            outputs = model(feat, time_t)
            all_z.append(outputs['latent_z'][0].cpu().numpy())

    z_concat = np.vstack(all_z)

    # t-SNE
    tsne = TSNE(n_components=2, perplexity=min(30, len(z_concat)-1),
                random_state=SEED, n_iter=1000)
    z_2d = tsne.fit_transform(z_concat)

    # Color by normalized time (wear progression)
    time_colors = np.linspace(0, 1, len(z_2d))
    scatter = ax.scatter(z_2d[:, 0], z_2d[:, 1], c=time_colors,
                        cmap='coolwarm', s=20, alpha=0.8)
    plt.colorbar(scatter, ax=ax, label='Normalized Time')

    ax.set_title(f'{cid.upper()} - Latent Space', fontweight='bold')
    ax.set_xlabel('t-SNE 1')
    ax.set_ylabel('t-SNE 2')
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Latent Degradation Space Visualization',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('latent_space_tsne.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 25: Physics Decomposition - Neural vs Physics Contribution

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]

    model.eval()
    physics_contrib = []
    total_rates = []

    with torch.no_grad():
        for i in range(NUM_CUTS):
            W = torch.tensor([[train_predictions[cid]['wear'][i]]],
                           dtype=torch.float32).to(DEVICE)
            t = torch.tensor([[i / NUM_CUTS]], dtype=torch.float32).to(DEVICE)

            phys_rate = model.ode_func.physics_rate(W, t)
            physics_contrib.append(phys_rate.item())

    physics_arr = np.array(physics_contrib)
    cuts = np.arange(1, NUM_CUTS + 1)

    # Estimate total rate from wear differences
    wear_pred = train_predictions[cid]['wear']
    total_rate_est = np.gradient(wear_pred)
    total_rate_est = np.maximum(total_rate_est, 0)

    neural_contrib = total_rate_est - physics_arr[:len(total_rate_est)]

    ax.fill_between(cuts[:len(physics_arr)], 0, physics_arr,
                    alpha=0.4, color='#2196F3', label='Physics Component')
    ax.fill_between(cuts[:len(neural_contrib)], physics_arr[:len(neural_contrib)],
                    physics_arr[:len(neural_contrib)] + np.maximum(neural_contrib, 0),
                    alpha=0.4, color='#F44336', label='Neural Residual')
    ax.plot(cuts[:len(total_rate_est)], total_rate_est,
            'k-', linewidth=1.5, label='Total Rate')

    ax.set_title(f'{cid.upper()} - Rate Decomposition', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Wear Rate dW/dt')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Physics vs Neural Wear Rate Decomposition',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('physics_neural_decomposition.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 26: Error Distribution Analysis

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for idx, cid in enumerate(TRAIN_CUTTERS):
    y_true = train_wear_normalized[cid]
    y_pred = train_predictions[cid]['wear']
    min_len = min(len(y_true), len(y_pred))

    errors = y_true[:min_len] - y_pred[:min_len]

    # Error histogram
    ax_hist = axes[0, idx]
    ax_hist.hist(errors, bins=40, color=colors[cid], alpha=0.7, edgecolor='black',
                density=True)
    ax_hist.axvline(0, color='red', linestyle='--', linewidth=1.5)
    ax_hist.set_title(f'{cid.upper()} - Error Distribution', fontweight='bold')
    ax_hist.set_xlabel('Prediction Error')
    ax_hist.set_ylabel('Density')
    ax_hist.text(0.05, 0.95,
                f'μ = {np.mean(errors):.4f}\nσ = {np.std(errors):.4f}',
                transform=ax_hist.transAxes, fontsize=10,
                verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    ax_hist.grid(True, alpha=0.3)

    # QQ-like: scatter actual vs predicted
    ax_qq = axes[1, idx]
    ax_qq.scatter(y_true[:min_len], y_pred[:min_len], s=10, alpha=0.5,
                 color=colors[cid])
    ax_qq.plot([0, 1], [0, 1], 'r--', linewidth=1.5)
    ax_qq.set_title(f'{cid.upper()} - Actual vs Predicted', fontweight='bold')
    ax_qq.set_xlabel('Actual Wear')
    ax_qq.set_ylabel('Predicted Wear')
    ax_qq.set_aspect('equal')
    ax_qq.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Error Analysis',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('error_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 27: Save Model & Generate Final Summary Report

# %%
# Save model
save_dir = "saved_models"
os.makedirs(save_dir, exist_ok=True)

torch.save({
    'model_state_dict': model.state_dict(),
    'feature_scaler_mean': feature_scaler.mean_,
    'feature_scaler_scale': feature_scaler.scale_,
    'wear_scaler_min': wear_scaler.min_,
    'wear_scaler_scale': wear_scaler.scale_,
    'hyperparameters': {
        'FEATURE_DIM': FEATURE_DIM,
        'HIDDEN_DIM': HIDDEN_DIM,
        'LATENT_DIM': LATENT_DIM,
        'SEQ_LEN': SEQ_LEN,
        'NUM_EPOCHS': NUM_EPOCHS,
        'LEARNING_RATE': LEARNING_RATE,
    },
    'training_history': history,
}, os.path.join(save_dir, 'pinn_nde_tool_wear_model.pth'))

print("Model saved to saved_models/pinn_nde_tool_wear_model.pth")

# ========= FINAL SUMMARY =========
print("\n" + "=" * 80)
print("=" * 80)
print("  PINN-NDE: Physics-Informed Neural Differential Equation Framework")
print("  for Tool Wear Degradation Modeling and RUL Prediction")
print("=" * 80)
print("=" * 80)

print(f"""
FRAMEWORK SUMMARY:
──────────────────
Architecture:  Feature Encoder → Bi-GRU Temporal → Neural ODE → Multi-Head Decoder
Physics:       Archard-Usui hybrid wear rate + monotonicity + irreversibility + energy
ODE Solver:    Euler (10-step) with learned physics parameters
Uncertainty:   Heteroscedastic aleatoric via learned variance
Outputs:       Total wear, flute-level wear, RUL, phase, latent physics

KEY NOVELTIES:
──────────────
1. Hybrid Neural Wear Rate ODE: dW/dt = f_physics(W,t) + f_neural(z,W,t)
2. Multi-Physics Loss with energy consistency constraints
3. Latent temperature & stress inference from sensor signals
4. Phase-aware degradation detection (initial/steady/accelerated)
5. Physics-constrained Bayesian uncertainty quantification
6. Discoverable wear law via learned ODE parameters

DISCOVERED WEAR LAW PARAMETERS:
────────────────────────────────
  dW/dt = {a:.4f} · exp({b:.4f} · W) · (1 + {c:.4f} · t) + neural_residual

MODEL STATISTICS:
─────────────────
  Parameters:      {total_params:,}
  Training epochs: {len(history['total'])}
  Final loss:      {history['total'][-1]:.6f}
  GPU Memory:      ~2-3 GB (GTX 1080 compatible)

FIGURES GENERATED:
──────────────────
  1. training_curves.png           - Training dynamics
  2. wear_prediction_training.png  - Training cutter predictions
  3. wear_prediction_test.png      - Test cutter generalization
  4. flute_wear_decomposition.png  - Flute-level analysis
  5. rul_prediction.png            - RUL predictions
  6. phase_transition_analysis.png - Wear phase detection
  7. latent_physics_inference.png  - Temperature/stress inference
  8. discovered_wear_law.png       - Learned physics parameters
  9. ablation_study.png            - Physics constraint contributions
  10. baseline_comparison.png      - Method comparison
  11. uncertainty_calibration.png  - Calibration analysis
  12. latent_space_tsne.png        - Latent representations
  13. physics_neural_decomposition.png - Rate decomposition
  14. error_analysis.png           - Error distributions
""")

print("=" * 80)
print("  Ready for MSSP / RESS submission!")
print("=" * 80)
```

```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
"""
================================================================================
PINN-NDE V2: Physics-Informed Neural Differential Equation Framework
for Tool Wear Degradation Modeling in Machining Systems
================================================================================
Authors: Muhammad Umar, Faisal Saleem, Jong-Myon Kim
Target: MSSP / RESS

V2 Fixes:
- Removed NLL uncertainty from main training loss (was causing negative loss)
- Deeper feature encoder with residual connections
- Stronger LSTM/CNN baselines (more capacity, proper training)
- Fixed MAPE computation (clip near-zero denominators)
- Ablation study with re-initialized models and longer training
- Physics curriculum starts from epoch 1 properly
- ODE solver with RK4 instead of Euler for accuracy
================================================================================
"""

# %% [markdown]
# # Cell 1: Environment Setup & Imports

# %%
import os, sys, glob, warnings, random, time, math
from pathlib import Path
from collections import OrderedDict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, TensorDataset, ConcatDataset
from torch.optim.lr_scheduler import CosineAnnealingWarmRestarts, OneCycleLR

from scipy import signal as scipy_signal
from scipy.stats import pearsonr, spearmanr
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

import pywt

warnings.filterwarnings("ignore")
plt.rcParams.update({
    'font.size': 12, 'axes.labelsize': 14, 'axes.titlesize': 14,
    'xtick.labelsize': 11, 'ytick.labelsize': 11, 'legend.fontsize': 11,
    'figure.dpi': 150, 'savefig.dpi': 300, 'savefig.bbox': 'tight',
    'font.family': 'serif', 'mathtext.fontset': 'cm',
})

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"Device: {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")

# %% [markdown]
# # Cell 2: Dataset Configuration & Loading

# %%
BASE = r"E:\Collaboration Work\With Farooq\phm dataset\PHM Challange 2010 Milling"
TRAIN_CUTTERS = ["c1", "c4", "c6"]
TEST_CUTTERS  = ["c2", "c3", "c5"]

SENSOR_COLS = ["Force_X", "Force_Y", "Force_Z",
               "Vibration_X", "Vibration_Y", "Vibration_Z", "AE_RMS"]
NUM_SENSORS = len(SENSOR_COLS)
SAMPLING_RATE = 50000
NUM_CUTS = 315
NUM_FLUTES = 3

# Hyperparameters
FEATURE_DIM = 84
HIDDEN_DIM = 256
LATENT_DIM = 128
NUM_EPOCHS = 500
BATCH_SIZE = 32
LEARNING_RATE = 5e-4
WEIGHT_DECAY = 1e-4
SEQ_LEN = 30
ODE_STEPS = 20


def load_cut_data(cutter_id, cut_num, base_path=BASE):
    filepath = os.path.join(base_path, cutter_id, cutter_id,
                            f"c_{cutter_id[1]}_{cut_num:03d}.csv")
    df = pd.read_csv(filepath, header=None)
    df.columns = SENSOR_COLS
    return df


def load_wear_data(cutter_id, base_path=BASE):
    filepath = os.path.join(base_path, cutter_id, f"{cutter_id}_wear.csv")
    df = pd.read_csv(filepath)
    return df


print("=" * 60)
print("DATA LOADING VERIFICATION")
print("=" * 60)
for cid in TRAIN_CUTTERS:
    wear = load_wear_data(cid)
    sample_cut = load_cut_data(cid, 1)
    print(f"\n{cid.upper()}: {wear.shape[0]} cuts, "
          f"Wear range: [{wear['flute_1'].min():.1f}, {wear['flute_1'].max():.1f}] μm")
    print(f"  Sample cut shape: {sample_cut.shape}, "
          f"Duration: {sample_cut.shape[0]/SAMPLING_RATE*1000:.1f} ms")

# %% [markdown]
# # Cell 3: Multi-Domain Feature Extraction (84 features per cut)

# %%
class MultiDomainFeatureExtractor:
    """12 features per sensor × 7 sensors = 84 features per cut."""

    def __init__(self, fs=SAMPLING_RATE):
        self.fs = fs

    def time_domain(self, x):
        rms = np.sqrt(np.mean(x**2))
        peak = np.max(np.abs(x))
        crest = peak / (rms + 1e-10)
        kurt = float(pd.Series(x).kurtosis())
        skew = float(pd.Series(x).skew())
        std = np.std(x)
        mean_abs = np.mean(np.abs(x))
        shape = rms / (mean_abs + 1e-10)
        return [rms, peak, crest, kurt, skew, std, mean_abs, shape]

    def freq_domain(self, x):
        n = len(x)
        fft_vals = np.abs(np.fft.rfft(x))
        freqs = np.fft.rfftfreq(n, d=1.0/self.fs)
        power = fft_vals**2
        total = np.sum(power) + 1e-10
        centroid = np.sum(freqs * power) / total
        bw = np.sqrt(np.sum((freqs - centroid)**2 * power) / total)
        return [centroid, bw]

    def wavelet_energy(self, x, wavelet='db4', level=4):
        coeffs = pywt.wavedec(x, wavelet, level=level)
        energies = [np.sum(c**2) for c in coeffs]
        total = sum(energies) + 1e-10
        return [energies[1]/total, energies[2]/total]

    def extract_channel(self, x):
        return self.time_domain(x) + self.freq_domain(x) + self.wavelet_energy(x)

    def extract_cut(self, cut_df):
        features = []
        for col in SENSOR_COLS:
            features.extend(self.extract_channel(cut_df[col].values))
        return np.array(features, dtype=np.float32)

    def extract_cutter(self, cutter_id, n_cuts=NUM_CUTS):
        print(f"  Extracting {cutter_id}...", end=" ")
        feats = []
        for i in range(1, n_cuts + 1):
            feats.append(self.extract_cut(load_cut_data(cutter_id, i)))
            if i % 100 == 0: print(f"{i}", end=" ")
        print("Done!")
        return np.array(feats)


extractor = MultiDomainFeatureExtractor()
print("\n" + "=" * 60)
print("FEATURE EXTRACTION")
print("=" * 60)

train_features = {}
train_wear = {}
for cid in TRAIN_CUTTERS:
    train_features[cid] = extractor.extract_cutter(cid)
    train_wear[cid] = load_wear_data(cid)

test_features = {}
for cid in TEST_CUTTERS:
    test_features[cid] = extractor.extract_cutter(cid)

print(f"\nFeatures per cut: {train_features['c1'].shape[1]}")

# %% [markdown]
# # Cell 4: Normalization & Target Preparation

# %%
# Feature normalization
all_train_feats = np.vstack([train_features[cid] for cid in TRAIN_CUTTERS])
feature_scaler = StandardScaler()
feature_scaler.fit(all_train_feats)

for cid in TRAIN_CUTTERS:
    train_features[cid] = feature_scaler.transform(train_features[cid])
for cid in TEST_CUTTERS:
    test_features[cid] = feature_scaler.transform(test_features[cid])

# Wear normalization (MinMax across all training wear)
wear_scaler = MinMaxScaler(feature_range=(0, 1))
all_wear = []
for cid in TRAIN_CUTTERS:
    w = train_wear[cid]
    all_wear.append(w[['flute_1', 'flute_2', 'flute_3']].values.flatten())
all_wear_arr = np.concatenate(all_wear).reshape(-1, 1)
wear_scaler.fit(all_wear_arr)

train_wear_norm = {}
train_flute_norm = {}
train_rul = {}

for cid in TRAIN_CUTTERS:
    w = train_wear[cid]
    total = w[['flute_1', 'flute_2', 'flute_3']].mean(axis=1).values.reshape(-1, 1)
    train_wear_norm[cid] = wear_scaler.transform(total).flatten().astype(np.float32)

    flute = w[['flute_1', 'flute_2', 'flute_3']].values
    train_flute_norm[cid] = wear_scaler.transform(
        flute.reshape(-1, 1)).reshape(-1, 3).astype(np.float32)

    n = len(total)
    train_rul[cid] = np.linspace(1.0, 0.0, n).astype(np.float32)

print("=" * 60)
print("DATA PREPARATION SUMMARY")
print("=" * 60)
for cid in TRAIN_CUTTERS:
    print(f"{cid}: Features {train_features[cid].shape}, "
          f"Wear [{train_wear_norm[cid].min():.3f}, {train_wear_norm[cid].max():.3f}]")

# %% [markdown]
# # Cell 5: Dataset & DataLoader

# %%
class WearDataset(Dataset):
    def __init__(self, features, wear, rul, flute_wear, seq_len=SEQ_LEN):
        self.samples = []
        n = len(features)
        for i in range(n - seq_len):
            self.samples.append((
                torch.FloatTensor(features[i:i+seq_len]),
                torch.FloatTensor(wear[i:i+seq_len]),
                torch.FloatTensor(rul[i:i+seq_len]),
                torch.FloatTensor(flute_wear[i:i+seq_len]),
                torch.FloatTensor(np.arange(i, i+seq_len, dtype=np.float32) / n),
            ))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return self.samples[idx]


train_datasets = []
for cid in TRAIN_CUTTERS:
    ds = WearDataset(train_features[cid], train_wear_norm[cid],
                     train_rul[cid], train_flute_norm[cid])
    train_datasets.append(ds)
    print(f"{cid}: {len(ds)} sequences")

combined_ds = ConcatDataset(train_datasets)
train_loader = DataLoader(combined_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=0, pin_memory=True, drop_last=True)
print(f"\nTotal sequences: {len(combined_ds)}, Batches/epoch: {len(train_loader)}")

# %% [markdown]
# # Cell 6: Neural ODE Components (RK4 Solver)

# %%
class WearRateODE(nn.Module):
    """
    Hybrid wear rate: dW/dt = f_physics(W, t) + f_neural(z, W, t)
    Physics: Archard-Usui inspired exponential wear model
    Neural: Residual correction for unknown tribological effects
    Constraint: dW/dt >= 0 (irreversible degradation)
    """
    def __init__(self, latent_dim, hidden_dim=128):
        super().__init__()
        self.latent_dim = latent_dim

        # Learnable physics parameters
        self.log_a = nn.Parameter(torch.tensor(-1.0))   # base rate
        self.log_b = nn.Parameter(torch.tensor(-2.0))   # wear acceleration
        self.log_c = nn.Parameter(torch.tensor(-1.0))   # time factor

        # Neural residual
        self.neural = nn.Sequential(
            nn.Linear(latent_dim + 2, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.GELU(),
            nn.Linear(hidden_dim // 2, 1),
        )

    def physics_rate(self, W, t):
        a = torch.exp(self.log_a)
        b = torch.exp(self.log_b)
        c = torch.exp(self.log_c)
        return a * torch.exp(b * W) * (1.0 + c * t)

    def forward(self, t_val, z, W):
        """
        Args:
            t_val: scalar time
            z: (batch, latent_dim) latent encoding
            W: (batch, 1) current wear
        Returns:
            dW/dt: (batch, 1)
        """
        t_in = t_val * torch.ones_like(W)
        phys = self.physics_rate(W, t_in)
        neural_in = torch.cat([z, W, t_in], dim=-1)
        neural = self.neural(neural_in)
        # Softplus ensures non-negative total rate
        dWdt = F.softplus(phys + neural, beta=3)
        return dWdt


class RK4Solver(nn.Module):
    """4th-order Runge-Kutta solver for Neural ODE."""
    def __init__(self, ode_func, n_steps=20):
        super().__init__()
        self.ode_func = ode_func
        self.n_steps = n_steps

    def forward(self, z, W0, t0, t1):
        dt = (t1 - t0) / self.n_steps
        W = W0
        for i in range(self.n_steps):
            t = t0 + i * dt
            k1 = self.ode_func(t, z, W)
            k2 = self.ode_func(t + 0.5*dt, z, W + 0.5*dt*k1)
            k3 = self.ode_func(t + 0.5*dt, z, W + 0.5*dt*k2)
            k4 = self.ode_func(t + dt, z, W + dt*k3)
            W = W + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        return W


print("Neural ODE: WearRateODE + RK4 Solver (20 steps)")

# %% [markdown]
# # Cell 7: PINN-NDE Model Architecture

# %%
class ResidualBlock(nn.Module):
    """Residual MLP block."""
    def __init__(self, dim, dropout=0.1):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim, dim),
            nn.LayerNorm(dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(dim, dim),
            nn.LayerNorm(dim),
        )
        self.act = nn.GELU()

    def forward(self, x):
        return self.act(x + self.net(x))


class PINN_NDE(nn.Module):
    """
    Physics-Informed Neural Differential Equation Framework.

    Pipeline:
    1. ResidualEncoder: features → latent z
    2. BiGRU: temporal context
    3. Neural ODE (RK4): physics-constrained wear evolution
    4. Multi-head decoder: wear, flute-wear, RUL
    5. Auxiliary heads: uncertainty, phase, latent physics
    """
    def __init__(self, input_dim=FEATURE_DIM, hidden_dim=HIDDEN_DIM,
                 latent_dim=LATENT_DIM, n_flutes=3, ode_steps=ODE_STEPS):
        super().__init__()
        self.latent_dim = latent_dim

        # === 1. Feature Encoder with Residual ===
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            ResidualBlock(hidden_dim, dropout=0.15),
            ResidualBlock(hidden_dim, dropout=0.15),
            nn.Linear(hidden_dim, latent_dim),
            nn.LayerNorm(latent_dim),
        )

        # === 2. Temporal Encoder ===
        self.gru = nn.GRU(latent_dim, hidden_dim // 2, num_layers=3,
                          batch_first=True, bidirectional=True, dropout=0.15)
        self.temporal_proj = nn.Sequential(
            nn.Linear(hidden_dim, latent_dim),
            nn.LayerNorm(latent_dim),
            nn.GELU(),
        )

        # === 3. Neural ODE ===
        self.ode_func = WearRateODE(latent_dim, hidden_dim // 2)
        self.ode_solver = RK4Solver(self.ode_func, n_steps=ode_steps)

        # Initial wear predictor
        self.w0_net = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 4),
            nn.GELU(),
            nn.Linear(hidden_dim // 4, 1),
            nn.Sigmoid(),
        )

        # === 4. Decoders ===
        dec_in = latent_dim + 1  # z + W_ode

        self.wear_dec = nn.Sequential(
            nn.Linear(dec_in, hidden_dim // 2),
            nn.GELU(),
            nn.Linear(hidden_dim // 2, hidden_dim // 4),
            nn.GELU(),
            nn.Linear(hidden_dim // 4, 1),
            nn.Sigmoid(),
        )

        self.flute_dec = nn.Sequential(
            nn.Linear(dec_in, hidden_dim // 2),
            nn.GELU(),
            nn.Linear(hidden_dim // 2, n_flutes),
            nn.Sigmoid(),
        )

        self.rul_dec = nn.Sequential(
            nn.Linear(dec_in, hidden_dim // 2),
            nn.GELU(),
            nn.Linear(hidden_dim // 2, 1),
            nn.Sigmoid(),
        )

        # === 5. Uncertainty (aleatoric) ===
        self.unc_head = nn.Sequential(
            nn.Linear(dec_in, hidden_dim // 4),
            nn.GELU(),
            nn.Linear(hidden_dim // 4, 2),  # log_var for wear and RUL
        )

        # === 6. Phase classifier ===
        self.phase_head = nn.Sequential(
            nn.Linear(dec_in, hidden_dim // 4),
            nn.GELU(),
            nn.Linear(hidden_dim // 4, 3),
        )

        # === 7. Latent physics ===
        self.physics_head = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 4),
            nn.GELU(),
            nn.Linear(hidden_dim // 4, 2),
            nn.Softplus(),
        )

    def forward(self, feat_seq, time_seq):
        B, T, _ = feat_seq.shape

        # Encode
        encoded = self.encoder(feat_seq)  # (B, T, latent)

        # Temporal
        gru_out, _ = self.gru(encoded)
        z = self.temporal_proj(gru_out)  # (B, T, latent)

        # ODE: step-by-step wear evolution
        W = self.w0_net(z[:, 0, :])  # (B, 1) initial wear
        wear_states = [W]

        for t in range(1, T):
            t0 = time_seq[:, t-1].mean().item()
            t1 = time_seq[:, t].mean().item()
            W = self.ode_solver(z[:, t-1, :], W, t0, t1)
            wear_states.append(W)

        W_ode = torch.stack(wear_states, dim=1)  # (B, T, 1)

        # Decode
        dec_in = torch.cat([z, W_ode], dim=-1)  # (B, T, latent+1)

        wear_pred = self.wear_dec(dec_in).squeeze(-1)      # (B, T)
        flute_pred = self.flute_dec(dec_in)                 # (B, T, 3)
        rul_pred = self.rul_dec(dec_in).squeeze(-1)         # (B, T)

        unc = self.unc_head(dec_in)
        phase_logits = self.phase_head(dec_in)
        lat_phys = self.physics_head(z)

        return {
            'wear_pred': wear_pred,
            'flute_pred': flute_pred,
            'rul_pred': rul_pred,
            'wear_ode': W_ode.squeeze(-1),
            'log_var_wear': unc[:, :, 0],
            'log_var_rul': unc[:, :, 1],
            'phase_logits': phase_logits,
            'latent_temperature': lat_phys[:, :, 0],
            'latent_stress': lat_phys[:, :, 1],
            'latent_z': z,
        }


model = PINN_NDE().to(DEVICE)
total_params = sum(p.numel() for p in model.parameters())
print(f"\n{'='*60}")
print(f"PINN-NDE V2 MODEL")
print(f"{'='*60}")
print(f"Parameters: {total_params:,} ({total_params*4/1e6:.2f} MB)")
print(f"{'='*60}")

# %% [markdown]
# # Cell 8: Physics-Informed Loss (V2 - Fixed)

# %%
class PhysicsLoss(nn.Module):
    """
    V2 loss: No NLL in main training (was causing negative loss).
    Clean separation of data fidelity + physics constraints.
    """
    def __init__(self):
        super().__init__()
        self.huber = nn.SmoothL1Loss()
        self.mse = nn.MSELoss()

        # Base weights (will be scaled by curriculum)
        self.w = {
            'wear': 10.0,
            'rul': 5.0,
            'flute': 3.0,
            'mono': 0.0,       # ramped up by curriculum
            'irreversible': 0.0,
            'smooth': 0.0,
            'ode_consist': 0.0,
        }

    def set_curriculum(self, epoch, warmup=30):
        """Curriculum: data-only first, then gradually add physics."""
        alpha = min(1.0, max(0.0, (epoch - warmup) / 50.0))
        self.w['mono'] = 20.0 * alpha
        self.w['irreversible'] = 50.0 * alpha
        self.w['smooth'] = 2.0 * alpha
        self.w['ode_consist'] = 10.0 * alpha

    def forward(self, out, tgt):
        L = {}

        # Data fidelity
        L['wear'] = self.huber(out['wear_pred'], tgt['wear'])
        L['rul'] = self.huber(out['rul_pred'], tgt['rul'])
        L['flute'] = self.huber(out['flute_pred'], tgt['flute_wear'])

        # Physics: monotonicity
        dw = out['wear_pred'][:, 1:] - out['wear_pred'][:, :-1]
        L['mono'] = torch.mean(F.relu(-dw) ** 2)

        # Physics: irreversibility on ODE state
        dw_ode = out['wear_ode'][:, 1:] - out['wear_ode'][:, :-1]
        L['irreversible'] = torch.mean(F.relu(-dw_ode) ** 2)

        # Physics: smoothness (2nd derivative penalty)
        d2w = dw[:, 1:] - dw[:, :-1]
        L['smooth'] = torch.mean(d2w ** 2)

        # ODE-decoder consistency
        L['ode_consist'] = self.mse(out['wear_pred'], out['wear_ode'])

        # Total
        total = sum(self.w[k] * L[k] for k in L)
        L['total'] = total
        return L


criterion = PhysicsLoss()
print("Physics Loss V2 initialized (curriculum-based)")
print("  Data phase (epochs 1-30): wear + rul + flute only")
print("  Physics ramp (epochs 30-80): mono + irrev + smooth + ode")

# %% [markdown]
# # Cell 9: Training Loop V2

# %%
def train_v2(model, loader, criterion, epochs=NUM_EPOCHS, lr=LEARNING_RATE, device=DEVICE):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY)

    total_steps = epochs * len(loader)
    scheduler = OneCycleLR(optimizer, max_lr=lr, total_steps=total_steps,
                           pct_start=0.1, anneal_strategy='cos',
                           div_factor=10, final_div_factor=100)

    hist = {'total': [], 'wear': [], 'rul': [], 'mono': [],
            'irreversible': [], 'ode_consist': [], 'lr': []}

    best_loss = float('inf')
    best_state = None
    patience_cnt = 0
    patience = 60

    print(f"\n{'='*70}")
    print(f"TRAINING V2 | Epochs: {epochs} | LR: {lr} | Batch: {BATCH_SIZE}")
    print(f"{'='*70}\n")

    for ep in range(1, epochs + 1):
        model.train()
        criterion.set_curriculum(ep)
        ep_loss = {k: 0.0 for k in hist if k != 'lr'}
        nb = 0

        for batch in loader:
            feat, wear, rul, flute, time_idx = [b.to(device) for b in batch]

            out = model(feat, time_idx)
            tgt = {'wear': wear, 'rul': rul, 'flute_wear': flute}
            losses = criterion(out, tgt)

            optimizer.zero_grad()
            losses['total'].backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            scheduler.step()

            for k in ep_loss:
                if k in losses:
                    ep_loss[k] += losses[k].item()
            nb += 1

        for k in ep_loss:
            ep_loss[k] /= max(nb, 1)
            hist[k].append(ep_loss[k])
        hist['lr'].append(optimizer.param_groups[0]['lr'])

        if ep_loss['total'] < best_loss:
            best_loss = ep_loss['total']
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience_cnt = 0
        else:
            patience_cnt += 1

        if ep % 20 == 0 or ep == 1:
            print(f"Ep {ep:03d}/{epochs} | Tot: {ep_loss['total']:.6f} | "
                  f"Wear: {ep_loss['wear']:.6f} | RUL: {ep_loss['rul']:.6f} | "
                  f"Mono: {ep_loss['mono']:.6f} | ODE: {ep_loss['ode_consist']:.6f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.2e}")

        if patience_cnt >= patience:
            print(f"\nEarly stopping at epoch {ep}")
            break

    if best_state:
        model.load_state_dict(best_state)
        print(f"\nBest model restored (loss: {best_loss:.6f})")

    return hist


history = train_v2(model, train_loader, criterion)

# %% [markdown]
# # Cell 10: Training Curves

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

plots = [
    ('total', 'Total Loss', 'b'),
    ('wear', 'Wear Loss', 'r'),
    ('rul', 'RUL Loss', 'g'),
    ('mono', 'Monotonicity Loss', 'm'),
    ('ode_consist', 'ODE Consistency', 'c'),
    ('lr', 'Learning Rate', 'k'),
]

for idx, (key, title, color) in enumerate(plots):
    ax = axes[idx // 3, idx % 3]
    data = history[key]
    ax.plot(data, f'{color}-', linewidth=1.2)
    ax.set_title(title, fontweight='bold')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Loss' if key != 'lr' else 'LR')
    if key != 'lr' and min(data) > 0:
        ax.set_yscale('log')
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE V2 Training Dynamics', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('fig01_training_curves.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 11: Inference Function

# %%
@torch.no_grad()
def predict_trajectory(model, features, seq_len=SEQ_LEN, device=DEVICE):
    model.eval()
    n = len(features)

    accum = {
        'wear': np.zeros(n), 'rul': np.zeros(n),
        'flute': np.zeros((n, 3)),
        'var_w': np.zeros(n), 'var_r': np.zeros(n),
        'phase': np.zeros((n, 3)),
        'temp': np.zeros(n), 'stress': np.zeros(n),
    }
    count = np.zeros(n)

    for i in range(n - seq_len):
        feat = torch.FloatTensor(features[i:i+seq_len]).unsqueeze(0).to(device)
        t = torch.FloatTensor(np.arange(i, i+seq_len, dtype=np.float32)/n).unsqueeze(0).to(device)

        out = model(feat, t)

        w = out['wear_pred'][0].cpu().numpy()
        r = out['rul_pred'][0].cpu().numpy()
        fl = out['flute_pred'][0].cpu().numpy()
        vw = torch.exp(out['log_var_wear'][0]).cpu().numpy()
        vr = torch.exp(out['log_var_rul'][0]).cpu().numpy()
        ph = F.softmax(out['phase_logits'][0], dim=-1).cpu().numpy()
        tp = out['latent_temperature'][0].cpu().numpy()
        st = out['latent_stress'][0].cpu().numpy()

        for j in range(seq_len):
            idx = i + j
            accum['wear'][idx] += w[j]
            accum['rul'][idx] += r[j]
            accum['flute'][idx] += fl[j]
            accum['var_w'][idx] += vw[j]
            accum['var_r'][idx] += vr[j]
            accum['phase'][idx] += ph[j]
            accum['temp'][idx] += tp[j]
            accum['stress'][idx] += st[j]
            count[idx] += 1

    count = np.maximum(count, 1)
    return {
        'wear': accum['wear'] / count,
        'rul': accum['rul'] / count,
        'flute': accum['flute'] / count[:, None],
        'wear_std': np.sqrt(accum['var_w'] / count),
        'rul_std': np.sqrt(accum['var_r'] / count),
        'phase': accum['phase'] / count[:, None],
        'temp': accum['temp'] / count,
        'stress': accum['stress'] / count,
    }


print("Inference on training cutters...")
train_preds = {}
for cid in TRAIN_CUTTERS:
    train_preds[cid] = predict_trajectory(model, train_features[cid])
    print(f"  {cid}: done")

print("\nInference on test cutters...")
test_preds = {}
for cid in TEST_CUTTERS:
    test_preds[cid] = predict_trajectory(model, test_features[cid])
    print(f"  {cid}: done")

# %% [markdown]
# # Cell 12: Metrics (Fixed MAPE)

# %%
def metrics(y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    # Fixed MAPE: clip denominator to avoid division by near-zero
    mape = np.mean(np.abs((y_true - y_pred) / np.maximum(np.abs(y_true), 0.01))) * 100
    pr, _ = pearsonr(y_true, y_pred)
    sr, _ = spearmanr(y_true, y_pred)
    mono = np.sum(np.diff(y_pred) >= -1e-6) / max(len(y_pred)-1, 1) * 100
    return {'RMSE': rmse, 'MAE': mae, 'R²': r2, 'MAPE(%)': mape,
            'Pearson': pr, 'Spearman': sr, 'Mono(%)': mono}


print("=" * 80)
print("TRAINING CUTTER EVALUATION")
print("=" * 80)

all_metrics = []
for cid in TRAIN_CUTTERS:
    y_t = train_wear_norm[cid]
    y_p = train_preds[cid]['wear']
    n = min(len(y_t), len(y_p))
    m = metrics(y_t[:n], y_p[:n])
    m['Cutter'] = cid.upper()
    all_metrics.append(m)
    print(f"\n{cid.upper()}:")
    for k, v in m.items():
        if k != 'Cutter': print(f"  {k:12s}: {v:.4f}")

print("\n" + "=" * 80)
print("TEST CUTTER PLAUSIBILITY")
print("=" * 80)
for cid in TEST_CUTTERS:
    p = test_preds[cid]['wear']
    mono = np.sum(np.diff(p) >= -1e-6) / (len(p)-1) * 100
    print(f"\n{cid.upper()}: range=[{p.min():.4f}, {p.max():.4f}], "
          f"Mono={mono:.1f}%, σ_mean={test_preds[cid]['wear_std'].mean():.4f}")

df_m = pd.DataFrame(all_metrics).set_index('Cutter')
print(f"\n{'='*80}\nSUMMARY\n{'='*80}")
print(df_m.to_string(float_format='%.4f'))

# %% [markdown]
# # Cell 13: Wear Prediction Plots (Training)

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
clrs = {'c1': '#2196F3', 'c4': '#E91E63', 'c6': '#4CAF50'}

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    yt = train_wear_norm[cid]
    yp = train_preds[cid]['wear']
    ys = train_preds[cid]['wear_std']
    n = min(len(yt), len(yp))
    cuts = np.arange(1, n+1)

    ax.plot(cuts, yt[:n], 'k-', lw=2, label='Ground Truth', zorder=3)
    ax.plot(cuts, yp[:n], '-', color=clrs[cid], lw=1.5, label='PINN-NDE', zorder=2)
    ax.fill_between(cuts, yp[:n]-2*ys[:n], yp[:n]+2*ys[:n],
                    alpha=0.2, color=clrs[cid], label='95% CI')

    m = metrics(yt[:n], yp[:n])
    ax.text(0.05, 0.95, f"RMSE={m['RMSE']:.4f}\nR²={m['R²']:.4f}\nMono={m['Mono(%)']:.1f}%",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    ax.set_title(f'Cutter {cid.upper()} (Train)', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(loc='lower right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Wear Prediction with Uncertainty Bounds',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig02_wear_train.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 14: Test Cutter Predictions

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
tclrs = {'c2': '#FF9800', 'c3': '#9C27B0', 'c5': '#00BCD4'}

for idx, cid in enumerate(TEST_CUTTERS):
    ax = axes[idx]
    yp = test_preds[cid]['wear']
    ys = test_preds[cid]['wear_std']
    cuts = np.arange(1, len(yp)+1)

    ax.plot(cuts, yp, '-', color=tclrs[cid], lw=1.5, label='PINN-NDE')
    ax.fill_between(cuts, yp-2*ys, yp+2*ys, alpha=0.2, color=tclrs[cid], label='95% CI')

    mono = np.sum(np.diff(yp) >= -1e-6) / (len(yp)-1) * 100
    ax.text(0.05, 0.95, f"Mono={mono:.1f}%\nσ={ys.mean():.4f}",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    ax.set_title(f'Cutter {cid.upper()} (Test)', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(loc='lower right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Generalization to Unseen Cutters',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig03_wear_test.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 15: Flute-Level Decomposition

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fc = ['#F44336', '#2196F3', '#4CAF50']

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    w = train_wear[cid]
    pf = train_preds[cid]['flute']
    n = min(len(w), len(pf))
    cuts = np.arange(1, n+1)

    for f in range(3):
        gt = wear_scaler.transform(w[f'flute_{f+1}'].values[:n].reshape(-1,1)).flatten()
        ax.plot(cuts, gt, '--', color=fc[f], lw=1.2, alpha=0.6, label=f'F{f+1} GT')
        ax.plot(cuts, pf[:n, f], '-', color=fc[f], lw=1.5, label=f'F{f+1} Pred')

    ax.set_title(f'{cid.upper()} - Flute Decomposition', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized Wear')
    ax.legend(fontsize=7, ncol=2, loc='upper left')
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Flute-Level Wear Decomposition',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig04_flute_decomp.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 16: RUL Prediction

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    yt = train_rul[cid]
    yp = train_preds[cid]['rul']
    ys = train_preds[cid]['rul_std']
    n = min(len(yt), len(yp))
    cuts = np.arange(1, n+1)

    ax.plot(cuts, yt[:n], 'k-', lw=2, label='Ground Truth')
    ax.plot(cuts, yp[:n], '-', color=clrs[cid], lw=1.5, label='PINN-NDE')
    ax.fill_between(cuts, yp[:n]-2*ys[:n], yp[:n]+2*ys[:n],
                    alpha=0.2, color=clrs[cid], label='95% CI')

    m = metrics(yt[:n], yp[:n])
    ax.text(0.55, 0.95, f"RMSE={m['RMSE']:.4f}\nR²={m['R²']:.4f}",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    ax.set_title(f'{cid.upper()} - RUL', fontweight='bold')
    ax.set_xlabel('Cut Number')
    ax.set_ylabel('Normalized RUL')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Remaining Useful Life Prediction',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig05_rul.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 17: Phase Transition Analysis

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
plabels = ['Initial', 'Steady', 'Accelerated']
pcolors = ['#4CAF50', '#2196F3', '#F44336']

for idx, cid in enumerate(TRAIN_CUTTERS):
    phase = train_preds[cid]['phase']
    yt = train_wear_norm[cid]
    yp = train_preds[cid]['wear']
    n = min(len(phase), len(yt), len(yp))
    cuts = np.arange(1, n+1)

    ax1 = axes[0, idx]
    for p in range(3):
        ax1.fill_between(cuts, 0, phase[:n, p], alpha=0.3, color=pcolors[p])
        ax1.plot(cuts, phase[:n, p], '-', color=pcolors[p], lw=1.5, label=plabels[p])
    ax1.set_title(f'{cid.upper()} - Phase Probabilities', fontweight='bold')
    ax1.set_xlabel('Cut Number')
    ax1.set_ylabel('Probability')
    ax1.legend(fontsize=8)
    ax1.set_ylim(0, 1.05)
    ax1.grid(True, alpha=0.3)

    ax2 = axes[1, idx]
    dom = np.argmax(phase[:n], axis=1)
    ax2.plot(cuts, yt[:n], 'k-', lw=2, label='Ground Truth')
    for p in range(3):
        mask = dom == p
        if np.any(mask):
            ax2.scatter(cuts[mask], yp[:n][mask], c=pcolors[p], s=10, alpha=0.7, label=plabels[p])
    ax2.set_title(f'{cid.upper()} - Phase-Colored Wear', fontweight='bold')
    ax2.set_xlabel('Cut Number')
    ax2.set_ylabel('Normalized Wear')
    ax2.legend(fontsize=8)
    ax2.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Wear Phase Transition Detection',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('fig06_phase.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 18: Latent Physics (Temperature & Stress)

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for idx, cid in enumerate(TRAIN_CUTTERS):
    temp = train_preds[cid]['temp']
    stress = train_preds[cid]['stress']
    wear = train_wear_norm[cid]
    n = min(len(temp), len(wear))
    cuts = np.arange(1, n+1)

    ax1 = axes[0, idx]
    ax1.plot(cuts, temp[:n], '-', color='#FF5722', lw=1.5)
    corr_t, _ = pearsonr(temp[:n], wear[:n])
    ax1.text(0.05, 0.95, f'ρ(T,W)={corr_t:.3f}', transform=ax1.transAxes,
             fontsize=10, va='top', bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    ax1.set_title(f'{cid.upper()} - Temperature Proxy', fontweight='bold')
    ax1.set_xlabel('Cut Number')
    ax1.set_ylabel('Temperature (a.u.)')
    ax1.grid(True, alpha=0.3)

    ax2 = axes[1, idx]
    ax2.plot(cuts, stress[:n], '-', color='#3F51B5', lw=1.5)
    corr_s, _ = pearsonr(stress[:n], wear[:n])
    ax2.text(0.05, 0.95, f'ρ(σ,W)={corr_s:.3f}', transform=ax2.transAxes,
             fontsize=10, va='top', bbox=dict(boxstyle='round', facecolor='lightcyan', alpha=0.8))
    ax2.set_title(f'{cid.upper()} - Stress Proxy', fontweight='bold')
    ax2.set_xlabel('Cut Number')
    ax2.set_ylabel('Contact Stress (a.u.)')
    ax2.grid(True, alpha=0.3)

plt.suptitle('PINN-NDE: Latent Physics Inference from Sensor Signals',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig07_latent_physics.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 19: Discovered Wear Law

# %%
with torch.no_grad():
    a = torch.exp(model.ode_func.log_a).item()
    b = torch.exp(model.ode_func.log_b).item()
    c = torch.exp(model.ode_func.log_c).item()

print("=" * 60)
print("DISCOVERED WEAR LAW")
print("=" * 60)
print(f"\n  dW/dt = {a:.6f} · exp({b:.6f} · W) · (1 + {c:.6f} · t) + f_neural(z,W,t)\n")
print(f"  a = {a:.6f}  (base wear rate)")
print(f"  b = {b:.6f}  (wear acceleration)")
print(f"  c = {c:.6f}  (time-dependent factor)")

# Plot physics rate surface
W_range = torch.linspace(0, 1, 100).to(DEVICE)
fig, ax = plt.subplots(figsize=(10, 6))
for t_val in [0.0, 0.25, 0.5, 0.75, 1.0]:
    with torch.no_grad():
        W_in = W_range.unsqueeze(-1)
        t_in = t_val * torch.ones_like(W_in)
        rate = model.ode_func.physics_rate(W_in, t_in)
    ax.plot(W_range.cpu().numpy(), rate.cpu().numpy().flatten(), lw=2, label=f't={t_val:.2f}')

ax.set_xlabel('Normalized Wear (W)')
ax.set_ylabel('Physics Wear Rate dW/dt')
ax.set_title(r'Discovered Wear Law: $\frac{dW}{dt} = a \cdot e^{bW} \cdot (1+ct)$',
             fontsize=14, fontweight='bold')
ax.legend(title='Normalized Time')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('fig08_wear_law.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 20: Ablation Study (V2 - Proper Re-initialization)

# %%
def train_ablation(config_name, mono_w, irrev_w, ode_w, smooth_w,
                   epochs=250, device=DEVICE):
    """Train ablation variant from scratch with proper init."""
    print(f"\n{'─'*50}")
    print(f"Ablation: {config_name}")
    print(f"  mono={mono_w}, irrev={irrev_w}, ode={ode_w}, smooth={smooth_w}")
    print(f"{'─'*50}")

    # Fresh model
    torch.manual_seed(SEED)
    abl_model = PINN_NDE().to(device)

    class AblLoss(nn.Module):
        def __init__(self, mw, iw, ow, sw):
            super().__init__()
            self.huber = nn.SmoothL1Loss()
            self.mse = nn.MSELoss()
            self.mw, self.iw, self.ow, self.sw = mw, iw, ow, sw

        def forward(self, out, tgt, epoch):
            L = {}
            L['wear'] = self.huber(out['wear_pred'], tgt['wear'])
            L['rul'] = self.huber(out['rul_pred'], tgt['rul'])
            L['flute'] = self.huber(out['flute_pred'], tgt['flute_wear'])

            alpha = min(1.0, max(0.0, (epoch - 30) / 50.0))

            dw = out['wear_pred'][:, 1:] - out['wear_pred'][:, :-1]
            L['mono'] = torch.mean(F.relu(-dw)**2)

            dw_ode = out['wear_ode'][:, 1:] - out['wear_ode'][:, :-1]
            L['irrev'] = torch.mean(F.relu(-dw_ode)**2)

            d2w = dw[:, 1:] - dw[:, :-1]
            L['smooth'] = torch.mean(d2w**2)

            L['ode'] = self.mse(out['wear_pred'], out['wear_ode'])

            total = (10*L['wear'] + 5*L['rul'] + 3*L['flute']
                     + self.mw*alpha*L['mono'] + self.iw*alpha*L['irrev']
                     + self.sw*alpha*L['smooth'] + self.ow*alpha*L['ode'])
            L['total'] = total
            return L

    abl_crit = AblLoss(mono_w, irrev_w, ode_w, smooth_w)
    opt = torch.optim.AdamW(abl_model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = epochs * len(train_loader)
    sched = OneCycleLR(opt, max_lr=LEARNING_RATE, total_steps=total_steps,
                       pct_start=0.1, anneal_strategy='cos', div_factor=10, final_div_factor=100)

    best_loss, best_state = float('inf'), None

    for ep in range(1, epochs+1):
        abl_model.train()
        tloss = 0; nb = 0
        for batch in train_loader:
            feat, wear, rul, flute, tidx = [b.to(device) for b in batch]
            out = abl_model(feat, tidx)
            losses = abl_crit(out, {'wear': wear, 'rul': rul, 'flute_wear': flute}, ep)
            opt.zero_grad()
            losses['total'].backward()
            torch.nn.utils.clip_grad_norm_(abl_model.parameters(), 1.0)
            opt.step()
            sched.step()
            tloss += losses['total'].item(); nb += 1

        avg = tloss / max(nb, 1)
        if avg < best_loss:
            best_loss = avg
            best_state = {k: v.cpu().clone() for k, v in abl_model.state_dict().items()}

        if ep % 80 == 0:
            print(f"  Ep {ep}: loss={avg:.6f}")

    if best_state:
        abl_model.load_state_dict(best_state)

    return abl_model


print("=" * 70)
print("ABLATION STUDY V2")
print("=" * 70)

abl_configs = OrderedDict({
    'Full PINN-NDE':       (20.0, 50.0, 10.0, 2.0),
    'w/o Monotonicity':    (0.0,  50.0, 10.0, 2.0),
    'w/o Irreversibility': (20.0, 0.0,  10.0, 2.0),
    'w/o ODE Consistency': (20.0, 50.0, 0.0,  2.0),
    'w/o Smoothness':      (20.0, 50.0, 10.0, 0.0),
    'Pure Data-Driven':    (0.0,  0.0,  0.0,  0.0),
})

abl_results = {}
for name, (mw, iw, ow, sw) in abl_configs.items():
    if name == 'Full PINN-NDE':
        abl_m = model
    else:
        abl_m = train_ablation(name, mw, iw, ow, sw, epochs=250)

    # Evaluate on ALL training cutters
    rmses, r2s, monos = [], [], []
    for cid in TRAIN_CUTTERS:
        p = predict_trajectory(abl_m, train_features[cid])
        yt = train_wear_norm[cid]
        n = min(len(yt), len(p['wear']))
        m = metrics(yt[:n], p['wear'][:n])
        rmses.append(m['RMSE']); r2s.append(m['R²']); monos.append(m['Mono(%)'])

    abl_results[name] = {
        'RMSE': np.mean(rmses), 'R²': np.mean(r2s), 'Mono(%)': np.mean(monos),
        'RMSE_std': np.std(rmses), 'R²_std': np.std(r2s),
    }
    print(f"\n  >> {name}: RMSE={np.mean(rmses):.4f}±{np.std(rmses):.4f}, "
          f"R²={np.mean(r2s):.4f}, Mono={np.mean(monos):.1f}%")

# Ablation table
print(f"\n{'='*80}")
print("ABLATION SUMMARY (Mean over C1, C4, C6)")
print(f"{'='*80}")
df_abl = pd.DataFrame(abl_results).T
print(df_abl.to_string(float_format='%.4f'))

# Ablation bar chart
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
names = list(abl_results.keys())
bar_c = ['#4CAF50', '#FF9800', '#F44336', '#9C27B0', '#00BCD4', '#607D8B']

axes[0].barh(names, [abl_results[n]['RMSE'] for n in names], color=bar_c, edgecolor='k', lw=0.5)
axes[0].set_xlabel('RMSE ↓'); axes[0].set_title('RMSE', fontweight='bold')
axes[0].grid(True, alpha=0.3, axis='x')

axes[1].barh(names, [abl_results[n]['R²'] for n in names], color=bar_c, edgecolor='k', lw=0.5)
axes[1].set_xlabel('R² ↑'); axes[1].set_title('R² Score', fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='x')

axes[2].barh(names, [abl_results[n]['Mono(%)'] for n in names], color=bar_c, edgecolor='k', lw=0.5)
axes[2].set_xlabel('Monotonicity % ↑'); axes[2].set_title('Monotonicity', fontweight='bold')
axes[2].grid(True, alpha=0.3, axis='x')

plt.suptitle('Ablation: Contribution of Physics Constraints',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig09_ablation.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 21: Baseline Comparisons (V2 - Stronger Baselines)

# %%
from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.svm import SVR

print("=" * 70)
print("BASELINE COMPARISONS V2")
print("=" * 70)

X_train = np.vstack([train_features[cid] for cid in TRAIN_CUTTERS])
y_train = np.concatenate([train_wear_norm[cid] for cid in TRAIN_CUTTERS])

baseline_res = OrderedDict()

# 1. Ridge
ridge = Ridge(alpha=1.0).fit(X_train, y_train)
# 2. Gradient Boosting
gb = GradientBoostingRegressor(n_estimators=300, max_depth=6, lr=0.05,
                                subsample=0.8, random_state=SEED).fit(X_train, y_train)
# 3. Random Forest
rf = RandomForestRegressor(n_estimators=300, max_depth=10, random_state=SEED).fit(X_train, y_train)
# 4. SVR
svr = SVR(kernel='rbf', C=10, gamma='scale').fit(X_train, y_train)

ml_models = {'Ridge': ridge, 'GBR': gb, 'RF': rf, 'SVR': svr}

for mname, mmod in ml_models.items():
    rmses, r2s, monos = [], [], []
    for cid in TRAIN_CUTTERS:
        pred = np.clip(mmod.predict(train_features[cid]), 0, 1)
        yt = train_wear_norm[cid]
        n = min(len(yt), len(pred))
        m = metrics(yt[:n], pred[:n])
        rmses.append(m['RMSE']); r2s.append(m['R²']); monos.append(m['Mono(%)'])
    baseline_res[mname] = {'RMSE': np.mean(rmses), 'R²': np.mean(r2s), 'Mono(%)': np.mean(monos)}
    print(f"\n{mname}: RMSE={np.mean(rmses):.4f}, R²={np.mean(r2s):.4f}, Mono={np.mean(monos):.1f}%")

# 5. LSTM (stronger)
print("\n--- LSTM (256 hidden, 3 layers) ---")

class StrongLSTM(nn.Module):
    def __init__(self, dim=FEATURE_DIM, hid=256):
        super().__init__()
        self.lstm = nn.LSTM(dim, hid, 3, batch_first=True, dropout=0.2)
        self.head = nn.Sequential(nn.Linear(hid, 64), nn.GELU(), nn.Linear(64, 1), nn.Sigmoid())

    def forward(self, x, _=None):
        o, _ = self.lstm(x)
        p = self.head(o).squeeze(-1)
        dummy = torch.zeros_like(p)
        return {'wear_pred': p, 'rul_pred': p,
                'flute_pred': p.unsqueeze(-1).expand(-1,-1,3),
                'wear_ode': p, 'log_var_wear': dummy, 'log_var_rul': dummy,
                'phase_logits': torch.zeros(p.shape[0],p.shape[1],3,device=p.device),
                'latent_temperature': dummy, 'latent_stress': dummy,
                'latent_z': torch.zeros(p.shape[0],p.shape[1],LATENT_DIM,device=p.device)}

torch.manual_seed(SEED)
lstm_m = StrongLSTM().to(DEVICE)
lstm_opt = torch.optim.AdamW(lstm_m.parameters(), lr=1e-3, weight_decay=1e-4)
lstm_sched = OneCycleLR(lstm_opt, max_lr=1e-3, total_steps=250*len(train_loader),
                        pct_start=0.1, div_factor=10, final_div_factor=100)
lstm_crit = nn.SmoothL1Loss()

for ep in range(1, 251):
    lstm_m.train()
    for batch in train_loader:
        feat, wear = batch[0].to(DEVICE), batch[1].to(DEVICE)
        pred = lstm_m(feat)['wear_pred']
        loss = lstm_crit(pred, wear)
        lstm_opt.zero_grad(); loss.backward()
        torch.nn.utils.clip_grad_norm_(lstm_m.parameters(), 1.0)
        lstm_opt.step(); lstm_sched.step()
    if ep % 80 == 0: print(f"  Ep {ep}")

rmses, r2s, monos = [], [], []
for cid in TRAIN_CUTTERS:
    p = predict_trajectory(lstm_m, train_features[cid])
    yt = train_wear_norm[cid]
    n = min(len(yt), len(p['wear']))
    m = metrics(yt[:n], p['wear'][:n])
    rmses.append(m['RMSE']); r2s.append(m['R²']); monos.append(m['Mono(%)'])
    print(f"  {cid}: RMSE={m['RMSE']:.4f}, R²={m['R²']:.4f}, Mono={m['Mono(%)']:.1f}%")
baseline_res['LSTM'] = {'RMSE': np.mean(rmses), 'R²': np.mean(r2s), 'Mono(%)': np.mean(monos)}

# 6. Transformer
print("\n--- Transformer (4 heads, 3 layers) ---")

class TransformerBaseline(nn.Module):
    def __init__(self, dim=FEATURE_DIM, hid=256, nhead=4, nlayers=3):
        super().__init__()
        self.proj = nn.Linear(dim, hid)
        self.pos = nn.Parameter(torch.randn(1, 500, hid) * 0.02)
        enc_layer = nn.TransformerEncoderLayer(d_model=hid, nhead=nhead,
                                               dim_feedforward=hid*2,
                                               dropout=0.15, batch_first=True,
                                               activation='gelu')
        self.transformer = nn.TransformerEncoder(enc_layer, nlayers)
        self.head = nn.Sequential(nn.Linear(hid, 64), nn.GELU(), nn.Linear(64, 1), nn.Sigmoid())

    def forward(self, x, _=None):
        B, T, _ = x.shape
        h = self.proj(x) + self.pos[:, :T, :]
        h = self.transformer(h)
        p = self.head(h).squeeze(-1)
        dummy = torch.zeros_like(p)
        return {'wear_pred': p, 'rul_pred': p,
                'flute_pred': p.unsqueeze(-1).expand(-1,-1,3),
                'wear_ode': p, 'log_var_wear': dummy, 'log_var_rul': dummy,
                'phase_logits': torch.zeros(B,T,3,device=p.device),
                'latent_temperature': dummy, 'latent_stress': dummy,
                'latent_z': torch.zeros(B,T,LATENT_DIM,device=p.device)}

torch.manual_seed(SEED)
tf_m = TransformerBaseline().to(DEVICE)
tf_opt = torch.optim.AdamW(tf_m.parameters(), lr=1e-3, weight_decay=1e-4)
tf_sched = OneCycleLR(tf_opt, max_lr=1e-3, total_steps=250*len(train_loader),
                      pct_start=0.1, div_factor=10, final_div_factor=100)

for ep in range(1, 251):
    tf_m.train()
    for batch in train_loader:
        feat, wear = batch[0].to(DEVICE), batch[1].to(DEVICE)
        pred = tf_m(feat)['wear_pred']
        loss = lstm_crit(pred, wear)
        tf_opt.zero_grad(); loss.backward()
        torch.nn.utils.clip_grad_norm_(tf_m.parameters(), 1.0)
        tf_opt.step(); tf_sched.step()
    if ep % 80 == 0: print(f"  Ep {ep}")

rmses, r2s, monos = [], [], []
for cid in TRAIN_CUTTERS:
    p = predict_trajectory(tf_m, train_features[cid])
    yt = train_wear_norm[cid]
    n = min(len(yt), len(p['wear']))
    m = metrics(yt[:n], p['wear'][:n])
    rmses.append(m['RMSE']); r2s.append(m['R²']); monos.append(m['Mono(%)'])
    print(f"  {cid}: RMSE={m['RMSE']:.4f}, R²={m['R²']:.4f}, Mono={m['Mono(%)']:.1f}%")
baseline_res['Transformer'] = {'RMSE': np.mean(rmses), 'R²': np.mean(r2s), 'Mono(%)': np.mean(monos)}

# Add ours
rmses, r2s, monos = [], [], []
for cid in TRAIN_CUTTERS:
    yt = train_wear_norm[cid]
    yp = train_preds[cid]['wear']
    n = min(len(yt), len(yp))
    m = metrics(yt[:n], yp[:n])
    rmses.append(m['RMSE']); r2s.append(m['R²']); monos.append(m['Mono(%)'])
baseline_res['PINN-NDE (Ours)'] = {'RMSE': np.mean(rmses), 'R²': np.mean(r2s), 'Mono(%)': np.mean(monos)}

print(f"\n{'='*80}")
print("FINAL COMPARISON TABLE (Mean over C1, C4, C6)")
print(f"{'='*80}")
df_comp = pd.DataFrame(baseline_res).T
print(df_comp.to_string(float_format='%.4f'))

# %% [markdown]
# # Cell 22: Comparison Visualization

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
methods = list(baseline_res.keys())
n_m = len(methods)
bcolors = plt.cm.Set2(np.linspace(0, 1, n_m))

# Highlight ours
highlight = [m == 'PINN-NDE (Ours)' for m in methods]
edge_widths = [2.5 if h else 0.5 for h in highlight]

for ax_idx, (metric_key, xlabel, title) in enumerate([
    ('RMSE', 'RMSE ↓', 'RMSE'), ('R²', 'R² ↑', 'R² Score'), ('Mono(%)', 'Mono % ↑', 'Monotonicity')
]):
    vals = [baseline_res[m][metric_key] for m in methods]
    bars = axes[ax_idx].barh(methods, vals, color=bcolors, edgecolor='black')
    for bar, ew, h in zip(bars, edge_widths, highlight):
        bar.set_edgecolor('red' if h else 'black')
        bar.set_linewidth(ew)
    for bar, val in zip(bars, vals):
        axes[ax_idx].text(bar.get_width()*1.01, bar.get_y()+bar.get_height()/2,
                         f'{val:.4f}', va='center', fontsize=9)
    axes[ax_idx].set_xlabel(xlabel)
    axes[ax_idx].set_title(title, fontweight='bold')
    axes[ax_idx].grid(True, alpha=0.3, axis='x')

plt.suptitle('PINN-NDE vs. Baselines (Mean over Training Cutters)',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig10_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 23: Uncertainty Calibration

# %%
from scipy.stats import norm

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    yt = train_wear_norm[cid]
    yp = train_preds[cid]['wear']
    ys = train_preds[cid]['wear_std']
    n = min(len(yt), len(yp))

    errors = np.abs(yt[:n] - yp[:n])
    stds = np.maximum(ys[:n], 1e-8)

    ks = np.linspace(0.1, 3.0, 30)
    observed = [np.mean(errors <= k*stds)*100 for k in ks]
    expected = [(norm.cdf(k) - norm.cdf(-k))*100 for k in ks]

    ax.plot(expected, observed, 'o-', color=clrs[cid], ms=4, lw=1.5, label='Observed')
    ax.plot([0,100], [0,100], 'k--', lw=1, label='Ideal')
    ax.set_xlabel('Expected Coverage (%)')
    ax.set_ylabel('Observed Coverage (%)')
    ax.set_title(f'{cid.upper()} - Calibration', fontweight='bold')
    ax.legend(); ax.grid(True, alpha=0.3)
    ax.set_xlim(0,100); ax.set_ylim(0,100); ax.set_aspect('equal')

plt.suptitle('Uncertainty Calibration', fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig11_calibration.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 24: t-SNE Latent Space

# %%
from sklearn.manifold import TSNE

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    model.eval()
    zs = []
    with torch.no_grad():
        for i in range(0, NUM_CUTS - SEQ_LEN, SEQ_LEN//2):
            end = min(i+SEQ_LEN, NUM_CUTS)
            if end - i < SEQ_LEN: continue
            f = torch.FloatTensor(train_features[cid][i:i+SEQ_LEN]).unsqueeze(0).to(DEVICE)
            t = torch.FloatTensor(np.arange(i,i+SEQ_LEN,dtype=np.float32)/NUM_CUTS).unsqueeze(0).to(DEVICE)
            o = model(f, t)
            zs.append(o['latent_z'][0].cpu().numpy())

    z_all = np.vstack(zs)
    perp = min(30, len(z_all)-1)
    z2d = TSNE(n_components=2, perplexity=perp, random_state=SEED, n_iter=1000).fit_transform(z_all)

    tc = np.linspace(0, 1, len(z2d))
    sc = ax.scatter(z2d[:,0], z2d[:,1], c=tc, cmap='coolwarm', s=20, alpha=0.8)
    plt.colorbar(sc, ax=ax, label='Time (norm)')
    ax.set_title(f'{cid.upper()} Latent Space', fontweight='bold')
    ax.set_xlabel('t-SNE 1'); ax.set_ylabel('t-SNE 2')
    ax.grid(True, alpha=0.3)

plt.suptitle('Latent Degradation State Visualization',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig12_tsne.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 25: Physics vs Neural Decomposition

# %%
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, cid in enumerate(TRAIN_CUTTERS):
    ax = axes[idx]
    model.eval()
    phys = []
    with torch.no_grad():
        for i in range(NUM_CUTS):
            W = torch.tensor([[train_preds[cid]['wear'][i]]], device=DEVICE)
            t = torch.tensor([[i/NUM_CUTS]], device=DEVICE)
            phys.append(model.ode_func.physics_rate(W, t).item())

    phys = np.array(phys)
    cuts = np.arange(1, NUM_CUTS+1)
    total = np.maximum(np.gradient(train_preds[cid]['wear']), 0)
    neural = total - phys[:len(total)]

    ax.fill_between(cuts[:len(phys)], 0, phys, alpha=0.4, color='#2196F3', label='Physics')
    ax.fill_between(cuts[:len(neural)], phys[:len(neural)],
                    phys[:len(neural)]+np.maximum(neural,0),
                    alpha=0.4, color='#F44336', label='Neural')
    ax.plot(cuts[:len(total)], total, 'k-', lw=1.5, label='Total Rate')
    ax.set_title(f'{cid.upper()} - Rate Decomposition', fontweight='bold')
    ax.set_xlabel('Cut Number'); ax.set_ylabel('dW/dt')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.3)

plt.suptitle('Physics vs Neural Wear Rate Decomposition',
             fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.savefig('fig13_decomposition.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 26: Error Analysis

# %%
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for idx, cid in enumerate(TRAIN_CUTTERS):
    yt = train_wear_norm[cid]
    yp = train_preds[cid]['wear']
    n = min(len(yt), len(yp))
    errs = yt[:n] - yp[:n]

    axes[0,idx].hist(errs, bins=40, color=clrs[cid], alpha=0.7, edgecolor='k', density=True)
    axes[0,idx].axvline(0, color='red', ls='--', lw=1.5)
    axes[0,idx].set_title(f'{cid.upper()} Error Distribution', fontweight='bold')
    axes[0,idx].set_xlabel('Error'); axes[0,idx].set_ylabel('Density')
    axes[0,idx].text(0.05, 0.95, f'μ={np.mean(errs):.4f}\nσ={np.std(errs):.4f}',
                     transform=axes[0,idx].transAxes, fontsize=10, va='top',
                     bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    axes[0,idx].grid(True, alpha=0.3)

    axes[1,idx].scatter(yt[:n], yp[:n], s=10, alpha=0.5, color=clrs[cid])
    axes[1,idx].plot([0,1],[0,1], 'r--', lw=1.5)
    axes[1,idx].set_title(f'{cid.upper()} Actual vs Predicted', fontweight='bold')
    axes[1,idx].set_xlabel('Actual'); axes[1,idx].set_ylabel('Predicted')
    axes[1,idx].set_aspect('equal'); axes[1,idx].grid(True, alpha=0.3)

plt.suptitle('Error Analysis', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('fig14_errors.png', dpi=300, bbox_inches='tight')
plt.show()

# %% [markdown]
# # Cell 27: Save & Final Report

# %%
os.makedirs("saved_models", exist_ok=True)
torch.save({
    'model_state_dict': model.state_dict(),
    'feature_scaler': {'mean': feature_scaler.mean_, 'scale': feature_scaler.scale_},
    'wear_scaler': {'min': wear_scaler.min_, 'scale': wear_scaler.scale_},
    'config': {'FEATURE_DIM': FEATURE_DIM, 'HIDDEN_DIM': HIDDEN_DIM,
               'LATENT_DIM': LATENT_DIM, 'SEQ_LEN': SEQ_LEN, 'ODE_STEPS': ODE_STEPS},
    'history': history,
}, "saved_models/pinn_nde_v2.pth")

print("Model saved.\n")

with torch.no_grad():
    a = torch.exp(model.ode_func.log_a).item()
    b = torch.exp(model.ode_func.log_b).item()
    c = torch.exp(model.ode_func.log_c).item()

print("=" * 80)
print("  PINN-NDE V2: Complete Results Summary")
print("=" * 80)
print(f"""
ARCHITECTURE: ResidualEncoder → 3-layer BiGRU → Neural ODE (RK4) → Multi-Head Decoder
PHYSICS: Archard-Usui hybrid + Monotonicity + Irreversibility + Smoothness + ODE Consistency
PARAMETERS: {total_params:,}

DISCOVERED WEAR LAW:
  dW/dt = {a:.6f} · exp({b:.6f} · W) · (1 + {c:.6f} · t) + f_neural(z, W, t)

TRAINING METRICS (Mean C1/C4/C6):
  RMSE:         {np.mean([metrics(train_wear_norm[c][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))], train_preds[c]['wear'][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))])['RMSE'] for c in TRAIN_CUTTERS]):.4f}
  R²:           {np.mean([metrics(train_wear_norm[c][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))], train_preds[c]['wear'][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))])['R²'] for c in TRAIN_CUTTERS]):.4f}
  Monotonicity: {np.mean([metrics(train_wear_norm[c][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))], train_preds[c]['wear'][:min(len(train_wear_norm[c]),len(train_preds[c]['wear']))])['Mono(%)'] for c in TRAIN_CUTTERS]):.1f}%

FIGURES GENERATED: fig01-fig14 (14 publication-quality figures)
""")

print("=" * 80)
print("  READY FOR MSSP / RESS SUBMISSION")
print("=" * 80)
```
